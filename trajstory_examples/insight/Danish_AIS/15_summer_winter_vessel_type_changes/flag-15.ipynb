### **Question 1**: How do summer and winter movement counts differ among ferry, fishing, and cargo traffic?

#### Analysis action

Derive season from timestamp and compare trajectory-level measurements by ship_type.

```python
import pandas as pd
import numpy as np
points=pd.read_csv(dataset_path); vessels=pd.read_csv(user_dataset_path)
points['timestamp']=pd.to_datetime(points['timestamp'])
points=points.merge(vessels,on='vessel_id')
points['season']=np.where(points.timestamp.dt.month.isin([6,7,8]),'summer','winter')
print(points.groupby(['season','ship_type']).trajectory_id.nunique())
# Display the ground-truth visualization for this question.
from IPython.display import Image, SVG, display
display(Image(filename="visualizations/seasonal_traffic_by_type.png"))
```

### Summary of Findings (Flag 15)

Ferry and fishing traffic increases sharply during summer and remains much lower during winter, while cargo traffic shows less seasonal variation. Generally, in the summer months, the mean speeds of different vessel types are higher than in the winter months. The consistency between COG and heading is higher in the summer months than in the winter months, especially for fishing vessels.
