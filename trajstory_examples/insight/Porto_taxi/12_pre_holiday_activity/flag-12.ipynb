### Summary of Findings (Flag 12)

Pre-holiday and special-day activity is disproportionately concentrated late in the evening, and those evening trips are usually directed north while ordinary-evening destinations are more evenly distributed across north, east, central, west, and south sectors. Pre-holiday evenings also contain a much larger medium/long-distance share. Street hails lead in the morning for both day classes, then decline after midday, especially on pre-holiday days.
