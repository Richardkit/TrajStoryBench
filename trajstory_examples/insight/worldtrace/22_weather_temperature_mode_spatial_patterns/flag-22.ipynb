### Summary of Findings (Flag 22)

Despite changes in weather and temperature, most trajectories remain concentrated within a limited set of repeatedly visited areas, indicating a relatively stable spatial mobility pattern. Adverse weather is associated with a shift toward car travel and more spatially compact non-car movement: walking and cycling become less common, and the observed non-car trips tend to cover shorter distances. A small number of walking or cycling trajectories during adverse weather extend substantially beyond the usual spatial range of non-car trips, standing out from the otherwise compact bad-weather mobility pattern.
