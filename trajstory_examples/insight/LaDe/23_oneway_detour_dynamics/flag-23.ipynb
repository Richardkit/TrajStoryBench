## LaDe 23: Oneway Detour Dynamics (Flag 23)

### Dataset Overview

Constructed LaDe-style courier data on archived Shanghai WGS84 OSM roads. Task IDs, schedules, movement speeds, delivery windows, AOIs and regions are synthetic extensions, not original LaDe observations. Native fclass, oneway, bridge and tunnel attributes are copied from OSM; absence of a tag is encoded no, not independent confirmation that a structure is absent. GPS retains all road vertices; spatial spacing is at most 22 m, stationary intervals at most 55 s. Each travelling row and its road_segment_id describe the immediately preceding GPS interval; stationary rows have no road attribution. Speed is distance divided by time, excluding service/waiting. Route directness ratio means travelled distance divided by straight-line task-endpoint distance. One-way share is distance weighted (0–1), not percent. Movement phases label ordered workday route periods, not fixed clock bins. Directions are checked but OSM turn restrictions are not modeled. Tunnel coordinates are constructed map interpolation, not authentic underground GPS fixes. All times are local Shanghai time. Pending tasks intentionally have no completion time or arrival leg; other task completions coincide with stationary GPS fixes. Service visits are recorded intervals at roadside task sites, not official AOI boundaries. Repositioning and return legs remain visible but are excluded where an insight concerns task-to-task legs. For schema field mean_value_of_one_way_road_ratio, 2.01 is the mean route-directness ratio of pure one-way delivery legs; it is not a percentage or an exposure share.

The primary table is loaded into `df`; the supporting GPS table is loaded into `user_df`.

Goal: Investigate how road-network attributes are reflected in the courier's movement dynamics.

Role: Last-mile logistics analyst


Category: oneway_detour_dynamics

### **Question 1**: At comparable task separations, does higher one-way-road exposure increase detour more than it changes average speed?

#### Analysis action

Interpret the constructed descriptive association, not an estimated causal effect.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

legs = pd.read_csv("movement_legs.csv")
visits = pd.read_csv("aoi_visits.csv")
extra = {}
extra["aois"] = pd.read_csv("aois.csv")


def evidence(case, index, df, user_df, legs, visits, extra):
    import json
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(8.4, 5.0), constrained_layout=True)
    moving = user_df[user_df.activity_state.eq("travelling")].copy()
    done = df[df.status.eq("completed")].copy()
    value = {}
    if case == 21:
        order = ["primary", "secondary", "tertiary", "residential"]
        moving = moving.merge(
            legs[["leg_id", "period"]], on="leg_id", validate="many_to_one"
        )
        if index == 1:
            summary = (
                moving.groupby("fclass")[["step_distance_km", "step_duration_s"]]
                .sum()
                .reindex(order)
            )
            summary["movement_speed_kmh"] = (
                summary.step_distance_km / summary.step_duration_s * 3600
            )
            summary["completions"] = done.delivery_fclass.value_counts().reindex(
                order, fill_value=0
            )
            summary["completions_per_km"] = (
                summary.completions / summary.step_distance_km
            )
            ax.bar(
                order,
                summary.movement_speed_kmh,
                color="#5c8ea7",
                alpha=0.8,
                label="Movement speed",
            )
            twin = ax.twinx()
            twin.plot(order, summary.completions_per_km, "o-", color="#bd7843")
            ax.set(ylabel="Distance / moving time (km/h)", xlabel="Native OSM fclass")
            twin.set_ylabel("Delivery completions per travelled km", color="#bd7843")
            value = {"road_class_summary": summary.to_dict("index")}
        elif index == 2:
            summary = (
                moving.groupby(["period", "fclass"])
                .step_distance_km.sum()
                .unstack(fill_value=0)
            )
            summary = summary.div(summary.sum(axis=1), axis=0).reindex(
                ["early", "late"]
            )
            summary[order].plot.bar(stacked=True, ax=ax, rot=0)
            ax.set(
                ylabel="Share of all moving distance",
                xlabel="First / second half of scheduled delivery stops",
            )
            tasks = done.merge(
                legs[["leg_id", "period"]], left_on="arrival_leg_id", right_on="leg_id"
            )
            value = {
                "distance_shares": summary.to_dict("index"),
                "completion_counts": tasks.period.value_counts().to_dict(),
            }
        else:
            # Smooth spatial AND temporal event kernel, independently measured
            # from task-completion positions/times rather than a generator label.
            frames = []
            for tid, q in moving.groupby("trajectory_id"):
                t = done[done.trajectory_id.eq(tid)]
                x = q[["lng", "lat"]].to_numpy() * [95.2, 111]
                y = t[["delivery_lng", "delivery_lat"]].to_numpy() * [95.2, 111]
                ds = np.linalg.norm(x[:, None] - y[None, :], axis=2)
                a = (
                    pd.to_datetime(q.gps_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                b = (
                    pd.to_datetime(t.completion_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                dt = abs(a[:, None] - b[None, :]) / 6e10
                q = q.copy()
                q["nearby_activity"] = (np.exp(-ds / 0.30) * np.exp(-dt / 30)).sum(
                    axis=1
                )
                frames.append(q)
            points = pd.concat(frames)
            values = {}
            for cls in order:
                q = points[points.fclass.eq(cls)]
                bins = pd.qcut(q.nearby_activity, 5, duplicates="drop")
                med = q.groupby(bins, observed=True)[
                    ["nearby_activity", "speed_kmh"]
                ].median()
                line = ax.plot(med.nearby_activity, med.speed_kmh, "o-", label=cls)[0]
                sample = q.sample(min(350, len(q)), random_state=17)
                ax.scatter(
                    sample.nearby_activity,
                    sample.speed_kmh,
                    s=6,
                    alpha=0.10,
                    color=line.get_color(),
                )
                values[cls] = {
                    "points": len(q),
                    "spearman": q.nearby_activity.corr(q.speed_kmh, method="spearman"),
                    "low_to_high_activity_speed_drop_kmh": (
                        med.speed_kmh.iloc[0] - med.speed_kmh.iloc[-1]
                    ),
                    "quintile_medians": med.to_dict("list"),
                }
            ax.set(
                xlabel=(
                    "Nearby completion intensity (300 m / 30 min exponential kernel)"
                ),
                ylabel="Median moving speed (km/h)",
            )
            ax.legend()
            value = {"within_class_evidence": values}
    elif case == 22:
        if index == 1:
            counts = (
                done.groupby(["service_date", "aoi_id"]).size().unstack(fill_value=0)
            )
            span = user_df.groupby("trajectory_id").gps_time.agg(["min", "max"])
            duration = (
                pd.to_datetime(span["max"], format="mixed")
                - pd.to_datetime(span["min"], format="mixed")
            ).dt.total_seconds()
            shares = (
                visits.groupby(["trajectory_id", "aoi_id"])
                .duration_s.sum()
                .unstack(fill_value=0)
                .div(duration, axis=0)
            )
            summary = pd.DataFrame(
                {
                    "mean_deliveries_per_day": counts.mean(),
                    "mean_work_time_share": shares.mean(),
                }
            )
            summary.plot.bar(
                y="mean_deliveries_per_day", ax=ax, rot=0, legend=False, color="#608ea4"
            )
            ax.set(
                ylabel="Completed deliveries per observed day", xlabel="Constructed AOI"
            )
            twin = ax.twinx()
            twin.plot(
                np.arange(len(summary)),
                summary.mean_work_time_share * 100,
                "o-",
                color="#bd7843",
            )
            twin.set_ylabel("Mean observed workday time (%)")
            value = {
                "aoi_summary": summary.to_dict("index"),
                "observed_days": counts.shape[0],
                "regular_destination_fclasses": sorted(
                    done[done.regular_aoi.eq("yes")].delivery_fclass.unique().tolist()
                ),
                "fixed_receipt_sites": (
                    extra["receipt_events"]
                    .groupby("aoi_id")[["lng", "lat"]]
                    .nunique()
                    .to_dict("index")
                ),
            }
        elif index == 2:
            q = legs[legs.purpose.eq("delivery")]
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.4), constrained_layout=True)
            ax = axes[0]
            ax.boxplot(
                [q[q.period.eq(p)].route_directness_ratio for p in ["early", "late"]],
                tick_labels=["Early", "Late"],
            )
            ax.set(
                xlabel="Delivery route phase",
                ylabel="Travelled / straight-line distance",
                title=(
                    "Later delivery routes are more direct\nLower ratio = less detour"
                ),
            )
            dates = sorted(q.service_date.unique())
            totals = moving.groupby("service_date")[
                ["step_distance_km", "step_duration_s"]
            ].sum()
            baseline = totals.step_distance_km / totals.step_duration_s * 3600
            late = q[q.period.eq("late")].copy()
            late["day_mean_movement_speed_kmh"] = late.service_date.map(baseline)
            late["difference_from_day_mean_kmh"] = (
                late.mean_speed_kmh - late.day_mean_movement_speed_kmh
            )
            above = late.difference_from_day_mean_kmh.gt(0)
            x = late.service_date.map({day: i for i, day in enumerate(dates)}).to_numpy(
                dtype=float
            )
            x += np.where(late.groupby("service_date").cumcount().eq(0), -0.12, 0.12)
            axes[1].axhline(
                0,
                color="#555555",
                linewidth=1,
                linestyle="--",
                label="Same as day mean",
            )
            for mask, color, label in [
                (above, "#3976a8", "Above day mean"),
                (~above, "#c77c3d", "At or below day mean"),
            ]:
                axes[1].scatter(
                    x[mask],
                    late.loc[mask, "difference_from_day_mean_kmh"],
                    color=color,
                    s=48,
                    label=f"{label}: {int(mask.sum())}/{len(late)} ({mask.mean():.1%})",
                )
            axes[1].set(
                xticks=np.arange(len(dates)),
                xticklabels=[day[5:] for day in dates],
                xlabel="Workday — each point is a delivery leg as time drew to a close",
                ylabel="Leg speed − day mean movement speed (km/h)",
                title=(
                    "As time drew to a close:\nMore direct does not guarantee"
                    " above-average speed"
                ),
            )
            axes[1].margins(y=0.30)
            axes[1].tick_params(axis="x", rotation=45, labelsize=8)
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "period_means": (
                    q.groupby("period")[["route_directness_ratio", "mean_speed_kmh"]]
                    .mean()
                    .to_dict("index")
                ),
                "day_mean_definition": (
                    "All moving GPS distance / all moving GPS time, including return"
                    " legs and excluding stops"
                ),
                "daily_mean_movement_speed_kmh": baseline.to_dict(),
                "late_delivery_legs": len(late),
                "late_legs_above_day_mean": int(above.sum()),
                "late_share_above_day_mean": float(above.mean()),
                "late_leg_comparison": (
                    late[
                        [
                            "leg_id",
                            "service_date",
                            "mean_speed_kmh",
                            "day_mean_movement_speed_kmh",
                            "difference_from_day_mean_kmh",
                        ]
                    ].to_dict("records")
                ),
            }
        elif index == 3:
            q = done.copy()
            stamp = pd.to_datetime(q.completion_time, format="mixed")
            q["completion_hour"] = (
                stamp.dt.hour + stamp.dt.minute / 60 + stamp.dt.second / 3600
            )
            q["window_hours"] = (
                pd.to_datetime(q.time_window_end) - pd.to_datetime(q.time_window_start)
            ).dt.total_seconds() / 3600
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13.5, 5.5), constrained_layout=True)
            ax = axes[0]
            groups = [
                q[q.regular_aoi.eq("no") & q.window_hours.lt(9)],
                q[q.regular_aoi.eq("no") & q.window_hours.ge(9)],
                q[q.regular_aoi.eq("yes") & q.window_hours.ge(9)],
            ]
            labels = [
                "Non-regular AOIs\nWindow <9 h",
                "Non-regular AOIs\nWindow ≥9 h",
                "Regular AOIs (A/B)\nWindow ≥9 h",
            ]
            colors = ["#4f846c", "#3976a8", "#c77c3d"]
            boxes = ax.boxplot(
                [g.completion_hour for g in groups],
                tick_labels=labels,
                patch_artist=True,
            )
            for box, color in zip(boxes["boxes"], colors):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, g in enumerate(groups, 1):
                median = g.completion_hour.median()
                h = int(median)
                minute = int(round((median - h) * 60))
                ax.text(
                    i,
                    17.8,
                    f"n={len(g)}\nMedian {h:02}:{minute:02}",
                    ha="center",
                    fontsize=9,
                )
            ax.set(
                ylabel="Task completion time (Shanghai)",
                title="Both non-regular window groups finish earlier",
                ylim=(7.5, 18.5),
            )
            ax.yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            ax.tick_params(axis="x", labelsize=8)
            for g, label, color, marker in zip(groups, labels, colors, ["o", "s", "^"]):
                axes[1].scatter(
                    g.window_hours,
                    g.completion_hour,
                    s=16,
                    alpha=0.45,
                    color=color,
                    marker=marker,
                    label=label.replace("\n", "; "),
                )
            axes[1].set(
                xlabel="Service-window duration (hours)",
                ylabel="Task completion time (Shanghai)",
                title="Wide windows alone do not explain later completion",
                ylim=(7.5, 19.5),
            )
            axes[1].yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(alpha=0.15)
            nvisits = visits.groupby(["service_date", "aoi_id"]).size()
            value = {
                "group_medians": (
                    q.groupby("regular_aoi")[["completion_hour", "window_hours"]]
                    .median()
                    .to_dict("index")
                ),
                "regular_aoi_visits_per_day": (
                    nvisits.unstack(fill_value=0)[["A", "B"]].to_dict("list")
                ),
                "irregular_by_window_median_hour": (
                    q[q.regular_aoi.eq("no")]
                    .assign(
                        window_group=lambda d: np.where(
                            d.window_hours < 9, "Narrower (<9h)", "Wide (9h+)"
                        )
                    )
                    .groupby("window_group")
                    .completion_hour.median()
                    .to_dict()
                ),
                "plotted_groups": {
                    label.replace("\n", "; "): {
                        "tasks": len(g),
                        "median_completion_hour": float(g.completion_hour.median()),
                        "window_hours_min": float(g.window_hours.min()),
                        "window_hours_max": float(g.window_hours.max()),
                    }
                    for label, g in zip(labels, groups)
                },
                "uncompleted_tasks_excluded": int(df.status.ne("completed").sum()),
            }
        else:
            irregular = df[df.regular_aoi.eq("no")]
            summary = (
                irregular.groupby(["service_date", "status"])
                .size()
                .unstack(fill_value=0)
            )
            summary.plot.bar(stacked=True, ax=ax, rot=50)
            ax.set(ylabel="Available irregular-AOI tasks", xlabel="Workday")
            pending = irregular[irregular.status.eq("pending")]
            dates = pending.service_date.unique().tolist()
            min_dist = {}
            for day in dates:
                gps = user_df[user_df.service_date.eq(day)]
                destinations = pending[pending.service_date.eq(day)][
                    ["aoi_id", "delivery_lng", "delivery_lat"]
                ].drop_duplicates()
                for r in destinations.itertuples():
                    ds = np.linalg.norm(
                        (
                            gps[["lng", "lat"]].to_numpy()
                            - [r.delivery_lng, r.delivery_lat]
                        )
                        * [95200, 111000],
                        axis=1,
                    )
                    min_dist[r.aoi_id] = float(ds.min())
            value = {
                "pending_dates": dates,
                "pending_tasks": len(pending),
                "minimum_GPS_distance_to_unvisited_AOI_m": min_dist,
                "anomaly_day_irregular_visit_count": int(
                    len(
                        visits[
                            visits.service_date.isin(dates)
                            & ~visits.aoi_id.isin(["A", "B"])
                        ]
                    )
                ),
            }
    elif case == 23:
        q = legs[legs.purpose.eq("delivery")].copy()
        # Initial departure is not between two completed task locations.
        q = q[q.groupby("trajectory_id").cumcount().gt(0)]
        if index == 1:
            q = q[q.period.ne("exception") & q.separation_km.between(0.8, 1.3)]
            q["oneway_group"] = np.where(
                q.oneway_share >= 0.55,
                "High (55%+)",
                np.where(q.oneway_share <= 0.25, "Low (0–25%)", "Middle"),
            )
            q = q[q.oneway_group.ne("Middle")]
            summary = q.groupby("oneway_group")[
                ["separation_km", "route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            high, low = summary.loc["High (55%+)"], summary.loc["Low (0–25%)"]
            relative = pd.Series(
                {
                    "Route directness": (
                        (high.route_directness_ratio / low.route_directness_ratio - 1)
                        * 100
                    ),
                    "Average speed": (
                        (high.mean_speed_kmh / low.mean_speed_kmh - 1) * 100
                    ),
                }
            )
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), constrained_layout=True)
            ax = axes[0]
            for name, s in q.groupby("oneway_group"):
                ax.scatter(
                    s.separation_km, s.route_directness_ratio, label=name, alpha=0.6
                )
            ax.set(
                xlabel="Task endpoint separation (km)",
                ylabel="Travelled / straight-line distance",
                title="Matched endpoint separations (0.8–1.3 km)",
            )
            ax.legend()
            bars = axes[1].bar(
                relative.index,
                relative.values,
                color=["#3976a8", "#c77c3d"],
                alpha=0.82,
            )
            axes[1].axhline(0, color="#666666", linewidth=1)
            axes[1].bar_label(
                bars,
                labels=[
                    (
                        f"+{relative.iloc[0]:.1f}%\n{low.route_directness_ratio:.2f} →"
                        f" {high.route_directness_ratio:.2f}"
                    ),
                    (
                        f"+{relative.iloc[1]:.1f}%\n{low.mean_speed_kmh:.1f} →"
                        f" {high.mean_speed_kmh:.1f} km/h"
                    ),
                ],
                padding=5,
                fontsize=10,
            )
            axes[1].set(
                ylabel="High one-way group relative to low group (%)",
                title="Average speed differs less than route directness",
                ylim=(0, max(relative) * 1.42),
            )
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "matched_separation_range_km": [0.8, 1.3],
                "group_means": summary.to_dict("index"),
                "high_vs_low_relative_difference_percent": relative.to_dict(),
                "within_period_group_means": (
                    q.groupby(["period", "oneway_group"])[
                        ["route_directness_ratio", "mean_speed_kmh"]
                    ]
                    .mean()
                    .reset_index()
                    .to_dict("records")
                ),
            }
        elif index == 2:
            means = (
                q[q.period.ne("exception")]
                .groupby("period")[["mean_speed_kmh", "route_directness_ratio"]]
                .mean()
                .reindex(["early", "midday", "late"])
            )
            stamp = pd.to_datetime(q.departure_time, format="mixed")
            q["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.floor(q.departure_hour.min() * 6) / 6,
                np.ceil(q.departure_hour.max() * 6) / 6 + 1 / 12,
                1 / 6,
            )
            windows = []
            for center in centers:
                subset = q[q.departure_hour.between(center - 0.5, center + 0.5)]
                windows.append(
                    {
                        "center_hour": float(center),
                        "legs": len(subset),
                        **{
                            column: (
                                float(subset[column].median())
                                if len(subset) >= 8
                                else None
                            )
                            for column in ["mean_speed_kmh", "route_directness_ratio"]
                        },
                    }
                )
            trend = pd.DataFrame(windows)
            plt.close(fig)
            fig, axes = plt.subplots(
                2, 1, figsize=(9.5, 7), sharex=True, constrained_layout=True
            )
            fig.suptitle("Speed and detour over actual departure time", fontsize=15)
            for axis, column, color, title in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "#3976a8",
                    "Movement speed of each task-to-task leg",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "#c77c3d",
                    "Route detour of the same legs at the same times",
                ),
            ]:
                normal = q[q.period.ne("exception")]
                exception = q[q.period.eq("exception")]
                axis.scatter(
                    normal.departure_hour,
                    normal[column],
                    s=22,
                    color=color,
                    alpha=0.45,
                    label=(
                        f"Task-to-task legs ({len(normal)};"
                        f" {normal.service_date.nunique()} days)"
                    ),
                )
                axis.scatter(
                    exception.departure_hour,
                    exception[column],
                    s=110,
                    color="#8c405f",
                    marker="*",
                    label="Exceptional leg (insight 4)",
                    zorder=4,
                )
                axis.plot(
                    trend.center_hour,
                    trend[column],
                    color=color,
                    linewidth=2.5,
                    label="60-min rolling median (≥8 legs)",
                )
                axis.set_title(title, fontsize=11)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(x=0.03, y=0.30)
                axis.legend(loc="upper left", fontsize=8, frameon=False)
            axes[0].set_ylabel("Mean movement speed (km/h)")
            axes[1].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            ticks = np.arange(
                np.floor(q.departure_hour.min()), np.ceil(q.departure_hour.max()) + 1
            )
            axes[1].set(
                xticks=ticks,
                xticklabels=[f"{int(h):02}:00" for h in ticks],
                xlabel="Actual leg departure time (Shanghai local time; days overlaid)",
            )
            value = {
                "period_means": means.to_dict("index"),
                "period_means_note": (
                    "Legacy phase aggregates retained for compatibility; not used to"
                    " draw this chart."
                ),
                "plotted_task_to_task_legs": len(q),
                "exception_legs_included": int(q.period.eq("exception").sum()),
                "rolling_window_minutes": 60,
                "minimum_legs_per_window": 8,
                "clock_time_rolling_medians": [
                    window for window in windows if window["legs"] >= 8
                ],
                "omitted_window_centers_hour": [
                    window["center_hour"] for window in windows if window["legs"] < 8
                ],
            }
        elif index == 3:
            p = (
                q[q.period.ne("exception")]
                .groupby(["service_date", "period"])
                .agg(
                    detour_sd=("route_directness_ratio", "std"),
                    speed_sd=("mean_speed_kmh", "std"),
                    transitions=("fclass_transitions", "sum"),
                    distance=("distance_km", "sum"),
                )
            )
            p["transitions_per_km"] = p.transitions / p.distance
            cut = p.transitions_per_km.median()
            p["transition_group"] = np.where(
                p.transitions_per_km > cut, "Frequent", "Less frequent"
            )
            pearson = p.transitions_per_km.corr(p.detour_sd)
            spearman = p.transitions_per_km.corr(p.detour_sd, method="spearman")
            group_means = p.groupby("transition_group")[
                ["detour_sd", "speed_sd", "transitions_per_km"]
            ].mean()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12.5, 5.5), constrained_layout=True)
            ax = axes[0]
            ax.scatter(
                p.transitions_per_km,
                p.detour_sd,
                color="#3976a8",
                s=43,
                alpha=0.72,
                label=f"Day × route phase (n={len(p)})",
            )
            slope, intercept = np.polyfit(p.transitions_per_km, p.detour_sd, 1)
            line = np.linspace(
                p.transitions_per_km.min(), p.transitions_per_km.max(), 100
            )
            ax.plot(
                line,
                slope * line + intercept,
                "--",
                color="#244d69",
                linewidth=2,
                label="Descriptive linear fit",
            )
            ax.axvline(
                cut,
                color="#777777",
                linestyle=":",
                linewidth=1.3,
                label=f"Median split: {cut:.2f}/km",
            )
            ax.set(
                xlabel="Native fclass transitions per travelled km",
                ylabel="Within-period route-ratio SD",
                title=(
                    "Detour variability increases with road-class switching\nPearson"
                    f" r={pearson:.2f}; Spearman ρ={spearman:.2f}"
                ),
            )
            ax.legend(loc="upper left", fontsize=8, frameon=False)
            ax.grid(alpha=0.15)
            order = ["Less frequent", "Frequent"]
            arrays = [p.loc[p.transition_group.eq(name), "speed_sd"] for name in order]
            boxes = axes[1].boxplot(
                arrays, tick_labels=order, patch_artist=True, showfliers=True
            )
            for box, color in zip(boxes["boxes"], ["#789bb0", "#d09a58"]):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, name in enumerate(order, 1):
                mean = group_means.loc[name, "speed_sd"]
                axes[1].scatter(
                    i,
                    mean,
                    marker="D",
                    s=45,
                    color="black",
                    zorder=4,
                    label="Mean" if i == 1 else None,
                )
                axes[1].annotate(
                    f"Mean {mean:.2f}",
                    (i, mean),
                    xytext=(8, 5),
                    textcoords="offset points",
                    fontsize=9,
                )
            axes[1].set(
                xlabel="Transition-frequency group (median split)",
                ylabel="Across-leg speed SD (km/h)",
                title=(
                    "Speed variability can remain similar\nIndividual periods still"
                    " vary"
                ),
            )
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "correlation_transition_frequency_detour_sd": pearson,
                "spearman_transition_frequency_detour_sd": spearman,
                "group_means": (
                    p.groupby("transition_group")[
                        ["detour_sd", "speed_sd", "transitions_per_km"]
                    ]
                    .mean()
                    .to_dict("index")
                ),
                "period_count": len(p),
            }
        else:
            normal = q[q.period.ne("exception")]
            exception = q[q.period.eq("exception")]
            pure = normal[np.isclose(normal.oneway_share, 1, atol=1e-10)]
            other = normal[~normal.index.isin(pure.index)]
            pure_mean = float(pure.route_directness_ratio.mean())
            ax.scatter(
                other.oneway_share * 100,
                other.route_directness_ratio,
                alpha=0.35,
                label="Other task legs",
            )
            ax.scatter(
                pure.oneway_share * 100,
                pure.route_directness_ratio,
                marker="D",
                s=55,
                color="#3976a8",
                label=f"Pure one-way routes (mean ratio {pure_mean:.2f})",
            )
            ax.scatter(
                exception.oneway_share * 100,
                exception.route_directness_ratio,
                marker="*",
                s=220,
                color="#c66643",
                label="Exceptional period",
            )
            ax.axhline(
                pure_mean, color="#3976a8", linestyle="--", linewidth=1.3, alpha=0.8
            )
            ax.set(
                xlabel="One-way-road distance (%)",
                ylabel="Travelled / straight-line distance",
            )
            ax.legend()
            value = {
                "exception": (
                    exception[
                        [
                            "leg_id",
                            "route_directness_ratio",
                            "oneway_share",
                            "mean_speed_kmh",
                            "distance_km",
                            "separation_km",
                        ]
                    ].to_dict("records")
                ),
                "normal_speed_10_90_percentiles": (
                    normal.mean_speed_kmh.quantile([0.1, 0.9]).tolist()
                ),
                "pure_oneway_reference": {
                    "definition": (
                        "delivery legs whose travelled distance is entirely on native"
                        " OSM one-way edges"
                    ),
                    "legs": int(len(pure)),
                    "mean_route_directness_ratio": pure_mean,
                    "route_directness_ratio_std": float(
                        pure.route_directness_ratio.std(ddof=0)
                    ),
                    "route_directness_ratios": pure.route_directness_ratio.tolist(),
                },
            }
    elif case == 24:
        if index == 1:
            q = legs[legs.purpose.str.startswith("road_condition")]
            q.boxplot(column="speed_std_kmh", by="purpose", ax=ax)
            fig.suptitle("")
            ax.set(
                xlabel="Road condition",
                ylabel="Short-term GPS speed SD (km/h)",
                title="Stable residential travel versus alternating road classes",
            )
            value = {
                "condition_means": (
                    q.groupby("purpose")[
                        [
                            "speed_std_kmh",
                            "residential_share",
                            "tertiary_share",
                            "fclass_transitions",
                        ]
                    ]
                    .mean()
                    .to_dict("index")
                )
            }
        else:
            q = extra["crossings"].merge(
                legs,
                left_on="crossing_leg_id",
                right_on="leg_id",
                validate="one_to_one",
            )
            low = moving[moving.speed_kmh < 12].groupby("leg_id").step_duration_s.sum()
            q["low_speed_seconds"] = q.leg_id.map(low).fillna(0)
            approach = legs.set_index("leg_id").mean_speed_kmh
            q["speed_drop_kmh"] = q.approach_leg_id.map(approach) - q.mean_speed_kmh
            if index == 2:
                # Fluctuation is within-crossing GPS speed SD, not variation
                # between the average speeds of different crossings.
                recomputed = moving.groupby("leg_id").speed_kmh.agg(
                    lambda s: float(np.std(s.to_numpy(), ddof=0))
                )
                np.testing.assert_allclose(
                    q.speed_std_kmh, q.leg_id.map(recomputed), atol=1e-8
                )
                plt.close(fig)
                fig, axes = plt.subplots(
                    1, 2, figsize=(12.5, 5.5), constrained_layout=True
                )
                ax = axes[0]
                structures = (
                    q[["structure_osm_way_id", "structure_type"]]
                    .drop_duplicates()
                    .sort_values("structure_type")
                )
                ticks = []
                labels = []
                for axis, column, title, ylabel in [
                    (
                        axes[0],
                        "low_speed_seconds",
                        "Longer low-speed time later in the workday",
                        "Time below 12 km/h per crossing (seconds)",
                    ),
                    (
                        axes[1],
                        "speed_std_kmh",
                        "Greater speed fluctuations later in the workday",
                        "Within-crossing GPS speed SD (km/h)",
                    ),
                ]:
                    for i, structure in enumerate(structures.itertuples()):
                        group = q[
                            q.structure_osm_way_id.eq(structure.structure_osm_way_id)
                        ]
                        paired = group.pivot(
                            index=["service_date", "approach_fclass"],
                            columns="period",
                            values=column,
                        )
                        assert paired[["early", "late"]].notna().all().all()
                        for row in paired.itertuples():
                            axis.plot(
                                [i * 3, i * 3 + 1],
                                [row.early, row.late],
                                color="#aab4bc",
                                alpha=0.5,
                                linewidth=0.8,
                            )
                        for j, (period, color) in enumerate(
                            [("early", "#3976a8"), ("late", "#c77c3d")]
                        ):
                            values = group.loc[group.period.eq(period), column]
                            axis.scatter(
                                np.full(len(values), i * 3 + j),
                                values,
                                s=24,
                                color=color,
                                alpha=0.7,
                                label=(
                                    period.capitalize() + " crossing"
                                    if i == 0
                                    else None
                                ),
                                zorder=3,
                            )
                            mean = values.mean()
                            axis.plot(
                                [i * 3 + j - 0.18, i * 3 + j + 0.18],
                                [mean, mean],
                                color="black",
                                linewidth=2,
                                label="Group mean" if i == 0 and j == 0 else None,
                                zorder=4,
                            )
                            axis.annotate(
                                f"{mean:.2f}",
                                (i * 3 + j, values.max()),
                                xytext=(0, 10),
                                textcoords="offset points",
                                ha="center",
                                fontweight="bold",
                                fontsize=10,
                            )
                        if axis is axes[0]:
                            ticks.extend([i * 3, i * 3 + 1])
                            labels.extend(
                                [
                                    f"Early\n{structure.structure_type.title()}",
                                    f"Late\n{structure.structure_type.title()}",
                                ]
                            )
                    axis.set(
                        xticks=ticks,
                        xticklabels=labels,
                        ylabel=ylabel,
                        title=title,
                        xlabel=(
                            "Paired within the same day, structure and approach class"
                        ),
                    )
                    axis.set_ylim(
                        0,
                        (
                            group[column].max() * 1.5
                            if column == "speed_std_kmh"
                            else q[column].max() * 1.5
                        ),
                    )
                    axis.legend(loc="upper left", fontsize=8, frameon=False)
                    axis.grid(axis="y", alpha=0.15)
                value = {
                    "structure_period_means": (
                        q.groupby(["structure_osm_way_id", "structure_type", "period"])[
                            ["low_speed_seconds", "speed_std_kmh", "mean_speed_kmh"]
                        ]
                        .mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                    "crossings": len(q),
                    "unique_structures": q.structure_osm_way_id.nunique(),
                }
            else:
                q.boxplot(column="speed_drop_kmh", by="approach_fclass", ax=ax)
                fig.suptitle("")
                ax.set(
                    xlabel="Dominant road class in last 800 m before structure",
                    ylabel="Approach minus crossing speed (km/h)",
                    title="Road approach conditions and crossing slowdowns",
                )
                value = {
                    "approach_means": (
                        q.groupby("approach_fclass")[
                            ["speed_drop_kmh", "approach_dominant_share"]
                        ]
                        .mean()
                        .to_dict("index")
                    ),
                    "within_structure_period_means": (
                        q.groupby(["structure_osm_way_id", "period", "approach_fclass"])
                        .speed_drop_kmh.mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                }
    elif case == 25:
        if index == 1:
            summary = legs.groupby("region_id")[
                ["route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            summary["delivery_volume"] = done.groupby("region_id").size()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12, 5.6), constrained_layout=True)
            ax = axes[0]
            order = ["Puxi_Jingan_West", "Puxi_Huangpu_East"]
            labels = [
                name + f'\n{int(summary.loc[name,"delivery_volume"])} deliveries'
                for name in order
            ]
            for axis, column, title, ylabel in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "Lower movement speed in Puxi_Jingan_West",
                    "Leg mean movement speed (km/h)",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "Smaller detour in Puxi_Jingan_West",
                    "Travelled / straight-line distance\nHigher = more detour",
                ),
            ]:
                arrays = [
                    legs.loc[legs.region_id.eq(name), column].to_numpy()
                    for name in order
                ]
                boxes = axis.boxplot(
                    arrays, tick_labels=labels, patch_artist=True, showfliers=True
                )
                for box, color in zip(boxes["boxes"], ["#3976a8", "#c77c3d"]):
                    box.set_facecolor(color)
                    box.set_alpha(0.35)
                for i, name in enumerate(order, 1):
                    mean = summary.loc[name, column]
                    axis.scatter(
                        i,
                        mean,
                        marker="D",
                        s=48,
                        color="black",
                        zorder=4,
                        label="Mean" if i == 1 else None,
                    )
                    axis.annotate(
                        f"Mean {mean:.2f}",
                        (i, mean),
                        xytext=(14, 5),
                        textcoords="offset points",
                        fontsize=9,
                        bbox=dict(
                            facecolor="white", alpha=0.8, edgecolor="none", pad=1
                        ),
                    )
                axis.set(title=title, ylabel=ylabel)
                axis.legend(loc="upper left", frameon=False, fontsize=9)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(y=0.18)
            difference = (
                summary.loc[order[1], "route_directness_ratio"]
                / summary.loc[order[0], "route_directness_ratio"]
                - 1
            )
            axes[1].set_title(
                f"Puxi_Huangpu_East: {difference:.1%} higher mean ratio", fontsize=11
            )
            axes[1].set_ylim(0.9, 3.25)
            outside = int(legs.route_directness_ratio.gt(3.25).sum())
            axes[1].text(
                0.03,
                0.88,
                f"Main view: 0.9–3.25\n{outside} higher outlier shown in inset",
                transform=axes[1].transAxes,
                fontsize=8,
                va="top",
            )
            inset = axes[1].inset_axes([0.07, 0.43, 0.37, 0.34])
            inset.boxplot(
                [
                    legs.loc[legs.region_id.eq(name), "route_directness_ratio"]
                    for name in order
                ],
                tick_labels=["Jingan_West", "Huangpu_East"],
                showfliers=True,
            )
            inset.set_title("Full range (all legs)", fontsize=8)
            inset.tick_params(labelsize=6)
            inset.grid(axis="y", alpha=0.15)
            value = {
                "region_means": summary.to_dict("index"),
                "region_role_mapping": (
                    extra["regions"][["region_id", "comparison_role"]].to_dict(
                        "records"
                    )
                ),
            }
        elif index == 2:
            summary = legs.groupby(["region_id", "period"])[
                ["route_directness_ratio", "primary_share", "residential_share"]
            ].mean()
            timed = legs.copy()
            stamp = pd.to_datetime(timed.departure_time, format="mixed")
            timed["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.ceil(timed.departure_hour.min() * 4) / 4,
                np.floor(timed.departure_hour.max() * 4) / 4 + 0.01,
                0.25,
            )
            curves = {}
            plt.close(fig)
            fig, ax = plt.subplots(figsize=(10, 6), constrained_layout=True)
            inset = ax.inset_axes([0.08, 0.48, 0.34, 0.26])
            for name, color in [
                ("Puxi_Jingan_West", "#3976a8"),
                ("Puxi_Huangpu_East", "#c77c3d"),
            ]:
                group = timed[timed.region_id.eq(name)]
                ax.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=15,
                    color=color,
                    alpha=0.25,
                )
                inset.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=5,
                    color=color,
                    alpha=0.4,
                )
                y = []
                records = []
                for center in centers:
                    window = group[
                        group.departure_hour.between(center - 0.5, center + 0.5)
                    ]
                    valid = len(window) >= 8 and window.service_date.nunique() >= 3
                    y.append(
                        float(window.route_directness_ratio.median())
                        if valid
                        else np.nan
                    )
                    if valid:
                        records.append(
                            dict(
                                center_hour=float(center),
                                median_route_ratio=y[-1],
                                legs=len(window),
                                days=int(window.service_date.nunique()),
                            )
                        )
                curves[name] = records
                ax.plot(
                    centers,
                    y,
                    "o-",
                    color=color,
                    linewidth=2.3,
                    markersize=4,
                    label=name,
                )
                inset.plot(centers, y, color=color, linewidth=1)
            ticks = np.arange(8, np.ceil(timed.departure_hour.max()) + 0.1, 0.5)
            labels = [f"{int(h):02}:{int(round((h%1)*60)):02}" for h in ticks]
            ax.set(
                xticks=ticks,
                xticklabels=labels,
                ylabel="Travelled / straight-line distance (higher = more detour)",
                xlabel="Actual departure time (Shanghai local time; 14 days overlaid)",
                title=(
                    "Regional detour trends over actual departure time\nDots:"
                    " individual legs; lines: 60-min rolling medians, every 15 min"
                ),
                ylim=(0.9, 3.25),
                xlim=(7.9, float(timed.departure_hour.max()) + 0.1),
            )
            inset.set(
                title="Full range (all legs)",
                xlim=ax.get_xlim(),
                ylim=(0.9, float(timed.route_directness_ratio.max()) + 0.5),
            )
            inset.title.set_fontsize(8)
            inset.tick_params(labelsize=7)
            inset.set_xticks([8, 10, 12], ["08:00", "10:00", "12:00"])
            ax.text(
                0.03,
                0.85,
                "Main view: ratio 0.9–3.25;"
                f" {int(timed.route_directness_ratio.gt(3.25).sum())} higher outlier in"
                " inset\nTrend requires ≥8 legs from ≥3 days per window; no"
                " interpolation",
                transform=ax.transAxes,
                fontsize=8,
                va="top",
            )
            ax.legend(loc="upper left", frameon=False)
            value = {
                "region_period_means": summary.reset_index().to_dict("records"),
                "period_means_note": (
                    "Legacy phase summary; plot uses actual departure times, not phase"
                    " labels."
                ),
                "rolling_window_minutes": 60,
                "evaluation_interval_minutes": 15,
                "minimum_legs": 8,
                "minimum_days": 3,
                "regional_clock_time_trends": curves,
            }
        else:
            values = {}
            plt.close(fig)
            fig, axes = plt.subplots(
                1,
                2,
                figsize=(12, 5.8),
                sharex=True,
                sharey=True,
                constrained_layout=True,
            )
            ax = axes[0]
            for axis, name, color in zip(
                axes, ["Puxi_Jingan_West", "Puxi_Huangpu_East"], ["#3976a8", "#c77c3d"]
            ):
                q = legs[legs.region_id.eq(name)]
                axis.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=24,
                    alpha=0.5,
                )
                values[name] = {
                    "pearson": q.oneway_share.corr(q.route_directness_ratio),
                    "spearman": q.oneway_share.corr(
                        q.route_directness_ratio, method="spearman"
                    ),
                    "legs": len(q),
                    "oneway_share_unique_values": q.oneway_share.nunique(),
                    "route_ratio_unique_values": q.route_directness_ratio.nunique(),
                }
                axis.set(
                    title=(
                        f'{name}\nPearson r = {values[name]["pearson"]:.2f}; Spearman ρ'
                        f' = {values[name]["spearman"]:.2f}'
                    ),
                    xlabel="One-way-road share of travelled distance (%)",
                    xlim=(-3, 103),
                    ylim=(0.9, 3.25),
                )
                axis.text(
                    0.03,
                    0.96,
                    f"n = {len(q)}; main view: ratio"
                    f" 0.9–3.25\n{int(q.route_directness_ratio.gt(3.25).sum())} higher"
                    " outlier(s); full range below",
                    transform=axis.transAxes,
                    va="top",
                    fontsize=8,
                )
                inset = axis.inset_axes([0.08, 0.47, 0.38, 0.34])
                inset.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=5,
                    alpha=0.5,
                )
                inset.set(
                    xlim=(-3, 103),
                    ylim=(0.9, float(legs.route_directness_ratio.max()) + 0.5),
                    title="Full range (all legs)",
                )
                inset.title.set_fontsize(8)
                inset.tick_params(labelsize=6)
                axis.grid(alpha=0.15)
            axes[0].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            axes[1].tick_params(labelleft=True)
            value = {"within_region_associations": values}
    ax.grid(alpha=0.15)
    return json.loads(json.dumps(value, default=lambda x: x.item()))


evidence_value = evidence(23, 1, df, user_df, legs, visits, extra)
print(evidence_value)
plt.close("all")
plt.close("all")
# Display the verified visualization once.
display(Image(filename="visualizations/insight_1.png"))
```

### **Question 2**: Do speed and detour vary differently over actual departure time, without preassigned workday phases?

#### Analysis action

Interpret the constructed descriptive association, not an estimated causal effect.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

legs = pd.read_csv("movement_legs.csv")
visits = pd.read_csv("aoi_visits.csv")
extra = {}
extra["aois"] = pd.read_csv("aois.csv")


def evidence(case, index, df, user_df, legs, visits, extra):
    import json
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(8.4, 5.0), constrained_layout=True)
    moving = user_df[user_df.activity_state.eq("travelling")].copy()
    done = df[df.status.eq("completed")].copy()
    value = {}
    if case == 21:
        order = ["primary", "secondary", "tertiary", "residential"]
        moving = moving.merge(
            legs[["leg_id", "period"]], on="leg_id", validate="many_to_one"
        )
        if index == 1:
            summary = (
                moving.groupby("fclass")[["step_distance_km", "step_duration_s"]]
                .sum()
                .reindex(order)
            )
            summary["movement_speed_kmh"] = (
                summary.step_distance_km / summary.step_duration_s * 3600
            )
            summary["completions"] = done.delivery_fclass.value_counts().reindex(
                order, fill_value=0
            )
            summary["completions_per_km"] = (
                summary.completions / summary.step_distance_km
            )
            ax.bar(
                order,
                summary.movement_speed_kmh,
                color="#5c8ea7",
                alpha=0.8,
                label="Movement speed",
            )
            twin = ax.twinx()
            twin.plot(order, summary.completions_per_km, "o-", color="#bd7843")
            ax.set(ylabel="Distance / moving time (km/h)", xlabel="Native OSM fclass")
            twin.set_ylabel("Delivery completions per travelled km", color="#bd7843")
            value = {"road_class_summary": summary.to_dict("index")}
        elif index == 2:
            summary = (
                moving.groupby(["period", "fclass"])
                .step_distance_km.sum()
                .unstack(fill_value=0)
            )
            summary = summary.div(summary.sum(axis=1), axis=0).reindex(
                ["early", "late"]
            )
            summary[order].plot.bar(stacked=True, ax=ax, rot=0)
            ax.set(
                ylabel="Share of all moving distance",
                xlabel="First / second half of scheduled delivery stops",
            )
            tasks = done.merge(
                legs[["leg_id", "period"]], left_on="arrival_leg_id", right_on="leg_id"
            )
            value = {
                "distance_shares": summary.to_dict("index"),
                "completion_counts": tasks.period.value_counts().to_dict(),
            }
        else:
            # Smooth spatial AND temporal event kernel, independently measured
            # from task-completion positions/times rather than a generator label.
            frames = []
            for tid, q in moving.groupby("trajectory_id"):
                t = done[done.trajectory_id.eq(tid)]
                x = q[["lng", "lat"]].to_numpy() * [95.2, 111]
                y = t[["delivery_lng", "delivery_lat"]].to_numpy() * [95.2, 111]
                ds = np.linalg.norm(x[:, None] - y[None, :], axis=2)
                a = (
                    pd.to_datetime(q.gps_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                b = (
                    pd.to_datetime(t.completion_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                dt = abs(a[:, None] - b[None, :]) / 6e10
                q = q.copy()
                q["nearby_activity"] = (np.exp(-ds / 0.30) * np.exp(-dt / 30)).sum(
                    axis=1
                )
                frames.append(q)
            points = pd.concat(frames)
            values = {}
            for cls in order:
                q = points[points.fclass.eq(cls)]
                bins = pd.qcut(q.nearby_activity, 5, duplicates="drop")
                med = q.groupby(bins, observed=True)[
                    ["nearby_activity", "speed_kmh"]
                ].median()
                line = ax.plot(med.nearby_activity, med.speed_kmh, "o-", label=cls)[0]
                sample = q.sample(min(350, len(q)), random_state=17)
                ax.scatter(
                    sample.nearby_activity,
                    sample.speed_kmh,
                    s=6,
                    alpha=0.10,
                    color=line.get_color(),
                )
                values[cls] = {
                    "points": len(q),
                    "spearman": q.nearby_activity.corr(q.speed_kmh, method="spearman"),
                    "low_to_high_activity_speed_drop_kmh": (
                        med.speed_kmh.iloc[0] - med.speed_kmh.iloc[-1]
                    ),
                    "quintile_medians": med.to_dict("list"),
                }
            ax.set(
                xlabel=(
                    "Nearby completion intensity (300 m / 30 min exponential kernel)"
                ),
                ylabel="Median moving speed (km/h)",
            )
            ax.legend()
            value = {"within_class_evidence": values}
    elif case == 22:
        if index == 1:
            counts = (
                done.groupby(["service_date", "aoi_id"]).size().unstack(fill_value=0)
            )
            span = user_df.groupby("trajectory_id").gps_time.agg(["min", "max"])
            duration = (
                pd.to_datetime(span["max"], format="mixed")
                - pd.to_datetime(span["min"], format="mixed")
            ).dt.total_seconds()
            shares = (
                visits.groupby(["trajectory_id", "aoi_id"])
                .duration_s.sum()
                .unstack(fill_value=0)
                .div(duration, axis=0)
            )
            summary = pd.DataFrame(
                {
                    "mean_deliveries_per_day": counts.mean(),
                    "mean_work_time_share": shares.mean(),
                }
            )
            summary.plot.bar(
                y="mean_deliveries_per_day", ax=ax, rot=0, legend=False, color="#608ea4"
            )
            ax.set(
                ylabel="Completed deliveries per observed day", xlabel="Constructed AOI"
            )
            twin = ax.twinx()
            twin.plot(
                np.arange(len(summary)),
                summary.mean_work_time_share * 100,
                "o-",
                color="#bd7843",
            )
            twin.set_ylabel("Mean observed workday time (%)")
            value = {
                "aoi_summary": summary.to_dict("index"),
                "observed_days": counts.shape[0],
                "regular_destination_fclasses": sorted(
                    done[done.regular_aoi.eq("yes")].delivery_fclass.unique().tolist()
                ),
                "fixed_receipt_sites": (
                    extra["receipt_events"]
                    .groupby("aoi_id")[["lng", "lat"]]
                    .nunique()
                    .to_dict("index")
                ),
            }
        elif index == 2:
            q = legs[legs.purpose.eq("delivery")]
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.4), constrained_layout=True)
            ax = axes[0]
            ax.boxplot(
                [q[q.period.eq(p)].route_directness_ratio for p in ["early", "late"]],
                tick_labels=["Early", "Late"],
            )
            ax.set(
                xlabel="Delivery route phase",
                ylabel="Travelled / straight-line distance",
                title=(
                    "Later delivery routes are more direct\nLower ratio = less detour"
                ),
            )
            dates = sorted(q.service_date.unique())
            totals = moving.groupby("service_date")[
                ["step_distance_km", "step_duration_s"]
            ].sum()
            baseline = totals.step_distance_km / totals.step_duration_s * 3600
            late = q[q.period.eq("late")].copy()
            late["day_mean_movement_speed_kmh"] = late.service_date.map(baseline)
            late["difference_from_day_mean_kmh"] = (
                late.mean_speed_kmh - late.day_mean_movement_speed_kmh
            )
            above = late.difference_from_day_mean_kmh.gt(0)
            x = late.service_date.map({day: i for i, day in enumerate(dates)}).to_numpy(
                dtype=float
            )
            x += np.where(late.groupby("service_date").cumcount().eq(0), -0.12, 0.12)
            axes[1].axhline(
                0,
                color="#555555",
                linewidth=1,
                linestyle="--",
                label="Same as day mean",
            )
            for mask, color, label in [
                (above, "#3976a8", "Above day mean"),
                (~above, "#c77c3d", "At or below day mean"),
            ]:
                axes[1].scatter(
                    x[mask],
                    late.loc[mask, "difference_from_day_mean_kmh"],
                    color=color,
                    s=48,
                    label=f"{label}: {int(mask.sum())}/{len(late)} ({mask.mean():.1%})",
                )
            axes[1].set(
                xticks=np.arange(len(dates)),
                xticklabels=[day[5:] for day in dates],
                xlabel="Workday — each point is a delivery leg as time drew to a close",
                ylabel="Leg speed − day mean movement speed (km/h)",
                title=(
                    "As time drew to a close:\nMore direct does not guarantee"
                    " above-average speed"
                ),
            )
            axes[1].margins(y=0.30)
            axes[1].tick_params(axis="x", rotation=45, labelsize=8)
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "period_means": (
                    q.groupby("period")[["route_directness_ratio", "mean_speed_kmh"]]
                    .mean()
                    .to_dict("index")
                ),
                "day_mean_definition": (
                    "All moving GPS distance / all moving GPS time, including return"
                    " legs and excluding stops"
                ),
                "daily_mean_movement_speed_kmh": baseline.to_dict(),
                "late_delivery_legs": len(late),
                "late_legs_above_day_mean": int(above.sum()),
                "late_share_above_day_mean": float(above.mean()),
                "late_leg_comparison": (
                    late[
                        [
                            "leg_id",
                            "service_date",
                            "mean_speed_kmh",
                            "day_mean_movement_speed_kmh",
                            "difference_from_day_mean_kmh",
                        ]
                    ].to_dict("records")
                ),
            }
        elif index == 3:
            q = done.copy()
            stamp = pd.to_datetime(q.completion_time, format="mixed")
            q["completion_hour"] = (
                stamp.dt.hour + stamp.dt.minute / 60 + stamp.dt.second / 3600
            )
            q["window_hours"] = (
                pd.to_datetime(q.time_window_end) - pd.to_datetime(q.time_window_start)
            ).dt.total_seconds() / 3600
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13.5, 5.5), constrained_layout=True)
            ax = axes[0]
            groups = [
                q[q.regular_aoi.eq("no") & q.window_hours.lt(9)],
                q[q.regular_aoi.eq("no") & q.window_hours.ge(9)],
                q[q.regular_aoi.eq("yes") & q.window_hours.ge(9)],
            ]
            labels = [
                "Non-regular AOIs\nWindow <9 h",
                "Non-regular AOIs\nWindow ≥9 h",
                "Regular AOIs (A/B)\nWindow ≥9 h",
            ]
            colors = ["#4f846c", "#3976a8", "#c77c3d"]
            boxes = ax.boxplot(
                [g.completion_hour for g in groups],
                tick_labels=labels,
                patch_artist=True,
            )
            for box, color in zip(boxes["boxes"], colors):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, g in enumerate(groups, 1):
                median = g.completion_hour.median()
                h = int(median)
                minute = int(round((median - h) * 60))
                ax.text(
                    i,
                    17.8,
                    f"n={len(g)}\nMedian {h:02}:{minute:02}",
                    ha="center",
                    fontsize=9,
                )
            ax.set(
                ylabel="Task completion time (Shanghai)",
                title="Both non-regular window groups finish earlier",
                ylim=(7.5, 18.5),
            )
            ax.yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            ax.tick_params(axis="x", labelsize=8)
            for g, label, color, marker in zip(groups, labels, colors, ["o", "s", "^"]):
                axes[1].scatter(
                    g.window_hours,
                    g.completion_hour,
                    s=16,
                    alpha=0.45,
                    color=color,
                    marker=marker,
                    label=label.replace("\n", "; "),
                )
            axes[1].set(
                xlabel="Service-window duration (hours)",
                ylabel="Task completion time (Shanghai)",
                title="Wide windows alone do not explain later completion",
                ylim=(7.5, 19.5),
            )
            axes[1].yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(alpha=0.15)
            nvisits = visits.groupby(["service_date", "aoi_id"]).size()
            value = {
                "group_medians": (
                    q.groupby("regular_aoi")[["completion_hour", "window_hours"]]
                    .median()
                    .to_dict("index")
                ),
                "regular_aoi_visits_per_day": (
                    nvisits.unstack(fill_value=0)[["A", "B"]].to_dict("list")
                ),
                "irregular_by_window_median_hour": (
                    q[q.regular_aoi.eq("no")]
                    .assign(
                        window_group=lambda d: np.where(
                            d.window_hours < 9, "Narrower (<9h)", "Wide (9h+)"
                        )
                    )
                    .groupby("window_group")
                    .completion_hour.median()
                    .to_dict()
                ),
                "plotted_groups": {
                    label.replace("\n", "; "): {
                        "tasks": len(g),
                        "median_completion_hour": float(g.completion_hour.median()),
                        "window_hours_min": float(g.window_hours.min()),
                        "window_hours_max": float(g.window_hours.max()),
                    }
                    for label, g in zip(labels, groups)
                },
                "uncompleted_tasks_excluded": int(df.status.ne("completed").sum()),
            }
        else:
            irregular = df[df.regular_aoi.eq("no")]
            summary = (
                irregular.groupby(["service_date", "status"])
                .size()
                .unstack(fill_value=0)
            )
            summary.plot.bar(stacked=True, ax=ax, rot=50)
            ax.set(ylabel="Available irregular-AOI tasks", xlabel="Workday")
            pending = irregular[irregular.status.eq("pending")]
            dates = pending.service_date.unique().tolist()
            min_dist = {}
            for day in dates:
                gps = user_df[user_df.service_date.eq(day)]
                destinations = pending[pending.service_date.eq(day)][
                    ["aoi_id", "delivery_lng", "delivery_lat"]
                ].drop_duplicates()
                for r in destinations.itertuples():
                    ds = np.linalg.norm(
                        (
                            gps[["lng", "lat"]].to_numpy()
                            - [r.delivery_lng, r.delivery_lat]
                        )
                        * [95200, 111000],
                        axis=1,
                    )
                    min_dist[r.aoi_id] = float(ds.min())
            value = {
                "pending_dates": dates,
                "pending_tasks": len(pending),
                "minimum_GPS_distance_to_unvisited_AOI_m": min_dist,
                "anomaly_day_irregular_visit_count": int(
                    len(
                        visits[
                            visits.service_date.isin(dates)
                            & ~visits.aoi_id.isin(["A", "B"])
                        ]
                    )
                ),
            }
    elif case == 23:
        q = legs[legs.purpose.eq("delivery")].copy()
        # Initial departure is not between two completed task locations.
        q = q[q.groupby("trajectory_id").cumcount().gt(0)]
        if index == 1:
            q = q[q.period.ne("exception") & q.separation_km.between(0.8, 1.3)]
            q["oneway_group"] = np.where(
                q.oneway_share >= 0.55,
                "High (55%+)",
                np.where(q.oneway_share <= 0.25, "Low (0–25%)", "Middle"),
            )
            q = q[q.oneway_group.ne("Middle")]
            summary = q.groupby("oneway_group")[
                ["separation_km", "route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            high, low = summary.loc["High (55%+)"], summary.loc["Low (0–25%)"]
            relative = pd.Series(
                {
                    "Route directness": (
                        (high.route_directness_ratio / low.route_directness_ratio - 1)
                        * 100
                    ),
                    "Average speed": (
                        (high.mean_speed_kmh / low.mean_speed_kmh - 1) * 100
                    ),
                }
            )
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), constrained_layout=True)
            ax = axes[0]
            for name, s in q.groupby("oneway_group"):
                ax.scatter(
                    s.separation_km, s.route_directness_ratio, label=name, alpha=0.6
                )
            ax.set(
                xlabel="Task endpoint separation (km)",
                ylabel="Travelled / straight-line distance",
                title="Matched endpoint separations (0.8–1.3 km)",
            )
            ax.legend()
            bars = axes[1].bar(
                relative.index,
                relative.values,
                color=["#3976a8", "#c77c3d"],
                alpha=0.82,
            )
            axes[1].axhline(0, color="#666666", linewidth=1)
            axes[1].bar_label(
                bars,
                labels=[
                    (
                        f"+{relative.iloc[0]:.1f}%\n{low.route_directness_ratio:.2f} →"
                        f" {high.route_directness_ratio:.2f}"
                    ),
                    (
                        f"+{relative.iloc[1]:.1f}%\n{low.mean_speed_kmh:.1f} →"
                        f" {high.mean_speed_kmh:.1f} km/h"
                    ),
                ],
                padding=5,
                fontsize=10,
            )
            axes[1].set(
                ylabel="High one-way group relative to low group (%)",
                title="Average speed differs less than route directness",
                ylim=(0, max(relative) * 1.42),
            )
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "matched_separation_range_km": [0.8, 1.3],
                "group_means": summary.to_dict("index"),
                "high_vs_low_relative_difference_percent": relative.to_dict(),
                "within_period_group_means": (
                    q.groupby(["period", "oneway_group"])[
                        ["route_directness_ratio", "mean_speed_kmh"]
                    ]
                    .mean()
                    .reset_index()
                    .to_dict("records")
                ),
            }
        elif index == 2:
            means = (
                q[q.period.ne("exception")]
                .groupby("period")[["mean_speed_kmh", "route_directness_ratio"]]
                .mean()
                .reindex(["early", "midday", "late"])
            )
            stamp = pd.to_datetime(q.departure_time, format="mixed")
            q["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.floor(q.departure_hour.min() * 6) / 6,
                np.ceil(q.departure_hour.max() * 6) / 6 + 1 / 12,
                1 / 6,
            )
            windows = []
            for center in centers:
                subset = q[q.departure_hour.between(center - 0.5, center + 0.5)]
                windows.append(
                    {
                        "center_hour": float(center),
                        "legs": len(subset),
                        **{
                            column: (
                                float(subset[column].median())
                                if len(subset) >= 8
                                else None
                            )
                            for column in ["mean_speed_kmh", "route_directness_ratio"]
                        },
                    }
                )
            trend = pd.DataFrame(windows)
            plt.close(fig)
            fig, axes = plt.subplots(
                2, 1, figsize=(9.5, 7), sharex=True, constrained_layout=True
            )
            fig.suptitle("Speed and detour over actual departure time", fontsize=15)
            for axis, column, color, title in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "#3976a8",
                    "Movement speed of each task-to-task leg",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "#c77c3d",
                    "Route detour of the same legs at the same times",
                ),
            ]:
                normal = q[q.period.ne("exception")]
                exception = q[q.period.eq("exception")]
                axis.scatter(
                    normal.departure_hour,
                    normal[column],
                    s=22,
                    color=color,
                    alpha=0.45,
                    label=(
                        f"Task-to-task legs ({len(normal)};"
                        f" {normal.service_date.nunique()} days)"
                    ),
                )
                axis.scatter(
                    exception.departure_hour,
                    exception[column],
                    s=110,
                    color="#8c405f",
                    marker="*",
                    label="Exceptional leg (insight 4)",
                    zorder=4,
                )
                axis.plot(
                    trend.center_hour,
                    trend[column],
                    color=color,
                    linewidth=2.5,
                    label="60-min rolling median (≥8 legs)",
                )
                axis.set_title(title, fontsize=11)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(x=0.03, y=0.30)
                axis.legend(loc="upper left", fontsize=8, frameon=False)
            axes[0].set_ylabel("Mean movement speed (km/h)")
            axes[1].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            ticks = np.arange(
                np.floor(q.departure_hour.min()), np.ceil(q.departure_hour.max()) + 1
            )
            axes[1].set(
                xticks=ticks,
                xticklabels=[f"{int(h):02}:00" for h in ticks],
                xlabel="Actual leg departure time (Shanghai local time; days overlaid)",
            )
            value = {
                "period_means": means.to_dict("index"),
                "period_means_note": (
                    "Legacy phase aggregates retained for compatibility; not used to"
                    " draw this chart."
                ),
                "plotted_task_to_task_legs": len(q),
                "exception_legs_included": int(q.period.eq("exception").sum()),
                "rolling_window_minutes": 60,
                "minimum_legs_per_window": 8,
                "clock_time_rolling_medians": [
                    window for window in windows if window["legs"] >= 8
                ],
                "omitted_window_centers_hour": [
                    window["center_hour"] for window in windows if window["legs"] < 8
                ],
            }
        elif index == 3:
            p = (
                q[q.period.ne("exception")]
                .groupby(["service_date", "period"])
                .agg(
                    detour_sd=("route_directness_ratio", "std"),
                    speed_sd=("mean_speed_kmh", "std"),
                    transitions=("fclass_transitions", "sum"),
                    distance=("distance_km", "sum"),
                )
            )
            p["transitions_per_km"] = p.transitions / p.distance
            cut = p.transitions_per_km.median()
            p["transition_group"] = np.where(
                p.transitions_per_km > cut, "Frequent", "Less frequent"
            )
            pearson = p.transitions_per_km.corr(p.detour_sd)
            spearman = p.transitions_per_km.corr(p.detour_sd, method="spearman")
            group_means = p.groupby("transition_group")[
                ["detour_sd", "speed_sd", "transitions_per_km"]
            ].mean()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12.5, 5.5), constrained_layout=True)
            ax = axes[0]
            ax.scatter(
                p.transitions_per_km,
                p.detour_sd,
                color="#3976a8",
                s=43,
                alpha=0.72,
                label=f"Day × route phase (n={len(p)})",
            )
            slope, intercept = np.polyfit(p.transitions_per_km, p.detour_sd, 1)
            line = np.linspace(
                p.transitions_per_km.min(), p.transitions_per_km.max(), 100
            )
            ax.plot(
                line,
                slope * line + intercept,
                "--",
                color="#244d69",
                linewidth=2,
                label="Descriptive linear fit",
            )
            ax.axvline(
                cut,
                color="#777777",
                linestyle=":",
                linewidth=1.3,
                label=f"Median split: {cut:.2f}/km",
            )
            ax.set(
                xlabel="Native fclass transitions per travelled km",
                ylabel="Within-period route-ratio SD",
                title=(
                    "Detour variability increases with road-class switching\nPearson"
                    f" r={pearson:.2f}; Spearman ρ={spearman:.2f}"
                ),
            )
            ax.legend(loc="upper left", fontsize=8, frameon=False)
            ax.grid(alpha=0.15)
            order = ["Less frequent", "Frequent"]
            arrays = [p.loc[p.transition_group.eq(name), "speed_sd"] for name in order]
            boxes = axes[1].boxplot(
                arrays, tick_labels=order, patch_artist=True, showfliers=True
            )
            for box, color in zip(boxes["boxes"], ["#789bb0", "#d09a58"]):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, name in enumerate(order, 1):
                mean = group_means.loc[name, "speed_sd"]
                axes[1].scatter(
                    i,
                    mean,
                    marker="D",
                    s=45,
                    color="black",
                    zorder=4,
                    label="Mean" if i == 1 else None,
                )
                axes[1].annotate(
                    f"Mean {mean:.2f}",
                    (i, mean),
                    xytext=(8, 5),
                    textcoords="offset points",
                    fontsize=9,
                )
            axes[1].set(
                xlabel="Transition-frequency group (median split)",
                ylabel="Across-leg speed SD (km/h)",
                title=(
                    "Speed variability can remain similar\nIndividual periods still"
                    " vary"
                ),
            )
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "correlation_transition_frequency_detour_sd": pearson,
                "spearman_transition_frequency_detour_sd": spearman,
                "group_means": (
                    p.groupby("transition_group")[
                        ["detour_sd", "speed_sd", "transitions_per_km"]
                    ]
                    .mean()
                    .to_dict("index")
                ),
                "period_count": len(p),
            }
        else:
            normal = q[q.period.ne("exception")]
            exception = q[q.period.eq("exception")]
            pure = normal[np.isclose(normal.oneway_share, 1, atol=1e-10)]
            other = normal[~normal.index.isin(pure.index)]
            pure_mean = float(pure.route_directness_ratio.mean())
            ax.scatter(
                other.oneway_share * 100,
                other.route_directness_ratio,
                alpha=0.35,
                label="Other task legs",
            )
            ax.scatter(
                pure.oneway_share * 100,
                pure.route_directness_ratio,
                marker="D",
                s=55,
                color="#3976a8",
                label=f"Pure one-way routes (mean ratio {pure_mean:.2f})",
            )
            ax.scatter(
                exception.oneway_share * 100,
                exception.route_directness_ratio,
                marker="*",
                s=220,
                color="#c66643",
                label="Exceptional period",
            )
            ax.axhline(
                pure_mean, color="#3976a8", linestyle="--", linewidth=1.3, alpha=0.8
            )
            ax.set(
                xlabel="One-way-road distance (%)",
                ylabel="Travelled / straight-line distance",
            )
            ax.legend()
            value = {
                "exception": (
                    exception[
                        [
                            "leg_id",
                            "route_directness_ratio",
                            "oneway_share",
                            "mean_speed_kmh",
                            "distance_km",
                            "separation_km",
                        ]
                    ].to_dict("records")
                ),
                "normal_speed_10_90_percentiles": (
                    normal.mean_speed_kmh.quantile([0.1, 0.9]).tolist()
                ),
                "pure_oneway_reference": {
                    "definition": (
                        "delivery legs whose travelled distance is entirely on native"
                        " OSM one-way edges"
                    ),
                    "legs": int(len(pure)),
                    "mean_route_directness_ratio": pure_mean,
                    "route_directness_ratio_std": float(
                        pure.route_directness_ratio.std(ddof=0)
                    ),
                    "route_directness_ratios": pure.route_directness_ratio.tolist(),
                },
            }
    elif case == 24:
        if index == 1:
            q = legs[legs.purpose.str.startswith("road_condition")]
            q.boxplot(column="speed_std_kmh", by="purpose", ax=ax)
            fig.suptitle("")
            ax.set(
                xlabel="Road condition",
                ylabel="Short-term GPS speed SD (km/h)",
                title="Stable residential travel versus alternating road classes",
            )
            value = {
                "condition_means": (
                    q.groupby("purpose")[
                        [
                            "speed_std_kmh",
                            "residential_share",
                            "tertiary_share",
                            "fclass_transitions",
                        ]
                    ]
                    .mean()
                    .to_dict("index")
                )
            }
        else:
            q = extra["crossings"].merge(
                legs,
                left_on="crossing_leg_id",
                right_on="leg_id",
                validate="one_to_one",
            )
            low = moving[moving.speed_kmh < 12].groupby("leg_id").step_duration_s.sum()
            q["low_speed_seconds"] = q.leg_id.map(low).fillna(0)
            approach = legs.set_index("leg_id").mean_speed_kmh
            q["speed_drop_kmh"] = q.approach_leg_id.map(approach) - q.mean_speed_kmh
            if index == 2:
                # Fluctuation is within-crossing GPS speed SD, not variation
                # between the average speeds of different crossings.
                recomputed = moving.groupby("leg_id").speed_kmh.agg(
                    lambda s: float(np.std(s.to_numpy(), ddof=0))
                )
                np.testing.assert_allclose(
                    q.speed_std_kmh, q.leg_id.map(recomputed), atol=1e-8
                )
                plt.close(fig)
                fig, axes = plt.subplots(
                    1, 2, figsize=(12.5, 5.5), constrained_layout=True
                )
                ax = axes[0]
                structures = (
                    q[["structure_osm_way_id", "structure_type"]]
                    .drop_duplicates()
                    .sort_values("structure_type")
                )
                ticks = []
                labels = []
                for axis, column, title, ylabel in [
                    (
                        axes[0],
                        "low_speed_seconds",
                        "Longer low-speed time later in the workday",
                        "Time below 12 km/h per crossing (seconds)",
                    ),
                    (
                        axes[1],
                        "speed_std_kmh",
                        "Greater speed fluctuations later in the workday",
                        "Within-crossing GPS speed SD (km/h)",
                    ),
                ]:
                    for i, structure in enumerate(structures.itertuples()):
                        group = q[
                            q.structure_osm_way_id.eq(structure.structure_osm_way_id)
                        ]
                        paired = group.pivot(
                            index=["service_date", "approach_fclass"],
                            columns="period",
                            values=column,
                        )
                        assert paired[["early", "late"]].notna().all().all()
                        for row in paired.itertuples():
                            axis.plot(
                                [i * 3, i * 3 + 1],
                                [row.early, row.late],
                                color="#aab4bc",
                                alpha=0.5,
                                linewidth=0.8,
                            )
                        for j, (period, color) in enumerate(
                            [("early", "#3976a8"), ("late", "#c77c3d")]
                        ):
                            values = group.loc[group.period.eq(period), column]
                            axis.scatter(
                                np.full(len(values), i * 3 + j),
                                values,
                                s=24,
                                color=color,
                                alpha=0.7,
                                label=(
                                    period.capitalize() + " crossing"
                                    if i == 0
                                    else None
                                ),
                                zorder=3,
                            )
                            mean = values.mean()
                            axis.plot(
                                [i * 3 + j - 0.18, i * 3 + j + 0.18],
                                [mean, mean],
                                color="black",
                                linewidth=2,
                                label="Group mean" if i == 0 and j == 0 else None,
                                zorder=4,
                            )
                            axis.annotate(
                                f"{mean:.2f}",
                                (i * 3 + j, values.max()),
                                xytext=(0, 10),
                                textcoords="offset points",
                                ha="center",
                                fontweight="bold",
                                fontsize=10,
                            )
                        if axis is axes[0]:
                            ticks.extend([i * 3, i * 3 + 1])
                            labels.extend(
                                [
                                    f"Early\n{structure.structure_type.title()}",
                                    f"Late\n{structure.structure_type.title()}",
                                ]
                            )
                    axis.set(
                        xticks=ticks,
                        xticklabels=labels,
                        ylabel=ylabel,
                        title=title,
                        xlabel=(
                            "Paired within the same day, structure and approach class"
                        ),
                    )
                    axis.set_ylim(
                        0,
                        (
                            group[column].max() * 1.5
                            if column == "speed_std_kmh"
                            else q[column].max() * 1.5
                        ),
                    )
                    axis.legend(loc="upper left", fontsize=8, frameon=False)
                    axis.grid(axis="y", alpha=0.15)
                value = {
                    "structure_period_means": (
                        q.groupby(["structure_osm_way_id", "structure_type", "period"])[
                            ["low_speed_seconds", "speed_std_kmh", "mean_speed_kmh"]
                        ]
                        .mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                    "crossings": len(q),
                    "unique_structures": q.structure_osm_way_id.nunique(),
                }
            else:
                q.boxplot(column="speed_drop_kmh", by="approach_fclass", ax=ax)
                fig.suptitle("")
                ax.set(
                    xlabel="Dominant road class in last 800 m before structure",
                    ylabel="Approach minus crossing speed (km/h)",
                    title="Road approach conditions and crossing slowdowns",
                )
                value = {
                    "approach_means": (
                        q.groupby("approach_fclass")[
                            ["speed_drop_kmh", "approach_dominant_share"]
                        ]
                        .mean()
                        .to_dict("index")
                    ),
                    "within_structure_period_means": (
                        q.groupby(["structure_osm_way_id", "period", "approach_fclass"])
                        .speed_drop_kmh.mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                }
    elif case == 25:
        if index == 1:
            summary = legs.groupby("region_id")[
                ["route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            summary["delivery_volume"] = done.groupby("region_id").size()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12, 5.6), constrained_layout=True)
            ax = axes[0]
            order = ["Puxi_Jingan_West", "Puxi_Huangpu_East"]
            labels = [
                name + f'\n{int(summary.loc[name,"delivery_volume"])} deliveries'
                for name in order
            ]
            for axis, column, title, ylabel in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "Lower movement speed in Puxi_Jingan_West",
                    "Leg mean movement speed (km/h)",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "Smaller detour in Puxi_Jingan_West",
                    "Travelled / straight-line distance\nHigher = more detour",
                ),
            ]:
                arrays = [
                    legs.loc[legs.region_id.eq(name), column].to_numpy()
                    for name in order
                ]
                boxes = axis.boxplot(
                    arrays, tick_labels=labels, patch_artist=True, showfliers=True
                )
                for box, color in zip(boxes["boxes"], ["#3976a8", "#c77c3d"]):
                    box.set_facecolor(color)
                    box.set_alpha(0.35)
                for i, name in enumerate(order, 1):
                    mean = summary.loc[name, column]
                    axis.scatter(
                        i,
                        mean,
                        marker="D",
                        s=48,
                        color="black",
                        zorder=4,
                        label="Mean" if i == 1 else None,
                    )
                    axis.annotate(
                        f"Mean {mean:.2f}",
                        (i, mean),
                        xytext=(14, 5),
                        textcoords="offset points",
                        fontsize=9,
                        bbox=dict(
                            facecolor="white", alpha=0.8, edgecolor="none", pad=1
                        ),
                    )
                axis.set(title=title, ylabel=ylabel)
                axis.legend(loc="upper left", frameon=False, fontsize=9)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(y=0.18)
            difference = (
                summary.loc[order[1], "route_directness_ratio"]
                / summary.loc[order[0], "route_directness_ratio"]
                - 1
            )
            axes[1].set_title(
                f"Puxi_Huangpu_East: {difference:.1%} higher mean ratio", fontsize=11
            )
            axes[1].set_ylim(0.9, 3.25)
            outside = int(legs.route_directness_ratio.gt(3.25).sum())
            axes[1].text(
                0.03,
                0.88,
                f"Main view: 0.9–3.25\n{outside} higher outlier shown in inset",
                transform=axes[1].transAxes,
                fontsize=8,
                va="top",
            )
            inset = axes[1].inset_axes([0.07, 0.43, 0.37, 0.34])
            inset.boxplot(
                [
                    legs.loc[legs.region_id.eq(name), "route_directness_ratio"]
                    for name in order
                ],
                tick_labels=["Jingan_West", "Huangpu_East"],
                showfliers=True,
            )
            inset.set_title("Full range (all legs)", fontsize=8)
            inset.tick_params(labelsize=6)
            inset.grid(axis="y", alpha=0.15)
            value = {
                "region_means": summary.to_dict("index"),
                "region_role_mapping": (
                    extra["regions"][["region_id", "comparison_role"]].to_dict(
                        "records"
                    )
                ),
            }
        elif index == 2:
            summary = legs.groupby(["region_id", "period"])[
                ["route_directness_ratio", "primary_share", "residential_share"]
            ].mean()
            timed = legs.copy()
            stamp = pd.to_datetime(timed.departure_time, format="mixed")
            timed["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.ceil(timed.departure_hour.min() * 4) / 4,
                np.floor(timed.departure_hour.max() * 4) / 4 + 0.01,
                0.25,
            )
            curves = {}
            plt.close(fig)
            fig, ax = plt.subplots(figsize=(10, 6), constrained_layout=True)
            inset = ax.inset_axes([0.08, 0.48, 0.34, 0.26])
            for name, color in [
                ("Puxi_Jingan_West", "#3976a8"),
                ("Puxi_Huangpu_East", "#c77c3d"),
            ]:
                group = timed[timed.region_id.eq(name)]
                ax.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=15,
                    color=color,
                    alpha=0.25,
                )
                inset.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=5,
                    color=color,
                    alpha=0.4,
                )
                y = []
                records = []
                for center in centers:
                    window = group[
                        group.departure_hour.between(center - 0.5, center + 0.5)
                    ]
                    valid = len(window) >= 8 and window.service_date.nunique() >= 3
                    y.append(
                        float(window.route_directness_ratio.median())
                        if valid
                        else np.nan
                    )
                    if valid:
                        records.append(
                            dict(
                                center_hour=float(center),
                                median_route_ratio=y[-1],
                                legs=len(window),
                                days=int(window.service_date.nunique()),
                            )
                        )
                curves[name] = records
                ax.plot(
                    centers,
                    y,
                    "o-",
                    color=color,
                    linewidth=2.3,
                    markersize=4,
                    label=name,
                )
                inset.plot(centers, y, color=color, linewidth=1)
            ticks = np.arange(8, np.ceil(timed.departure_hour.max()) + 0.1, 0.5)
            labels = [f"{int(h):02}:{int(round((h%1)*60)):02}" for h in ticks]
            ax.set(
                xticks=ticks,
                xticklabels=labels,
                ylabel="Travelled / straight-line distance (higher = more detour)",
                xlabel="Actual departure time (Shanghai local time; 14 days overlaid)",
                title=(
                    "Regional detour trends over actual departure time\nDots:"
                    " individual legs; lines: 60-min rolling medians, every 15 min"
                ),
                ylim=(0.9, 3.25),
                xlim=(7.9, float(timed.departure_hour.max()) + 0.1),
            )
            inset.set(
                title="Full range (all legs)",
                xlim=ax.get_xlim(),
                ylim=(0.9, float(timed.route_directness_ratio.max()) + 0.5),
            )
            inset.title.set_fontsize(8)
            inset.tick_params(labelsize=7)
            inset.set_xticks([8, 10, 12], ["08:00", "10:00", "12:00"])
            ax.text(
                0.03,
                0.85,
                "Main view: ratio 0.9–3.25;"
                f" {int(timed.route_directness_ratio.gt(3.25).sum())} higher outlier in"
                " inset\nTrend requires ≥8 legs from ≥3 days per window; no"
                " interpolation",
                transform=ax.transAxes,
                fontsize=8,
                va="top",
            )
            ax.legend(loc="upper left", frameon=False)
            value = {
                "region_period_means": summary.reset_index().to_dict("records"),
                "period_means_note": (
                    "Legacy phase summary; plot uses actual departure times, not phase"
                    " labels."
                ),
                "rolling_window_minutes": 60,
                "evaluation_interval_minutes": 15,
                "minimum_legs": 8,
                "minimum_days": 3,
                "regional_clock_time_trends": curves,
            }
        else:
            values = {}
            plt.close(fig)
            fig, axes = plt.subplots(
                1,
                2,
                figsize=(12, 5.8),
                sharex=True,
                sharey=True,
                constrained_layout=True,
            )
            ax = axes[0]
            for axis, name, color in zip(
                axes, ["Puxi_Jingan_West", "Puxi_Huangpu_East"], ["#3976a8", "#c77c3d"]
            ):
                q = legs[legs.region_id.eq(name)]
                axis.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=24,
                    alpha=0.5,
                )
                values[name] = {
                    "pearson": q.oneway_share.corr(q.route_directness_ratio),
                    "spearman": q.oneway_share.corr(
                        q.route_directness_ratio, method="spearman"
                    ),
                    "legs": len(q),
                    "oneway_share_unique_values": q.oneway_share.nunique(),
                    "route_ratio_unique_values": q.route_directness_ratio.nunique(),
                }
                axis.set(
                    title=(
                        f'{name}\nPearson r = {values[name]["pearson"]:.2f}; Spearman ρ'
                        f' = {values[name]["spearman"]:.2f}'
                    ),
                    xlabel="One-way-road share of travelled distance (%)",
                    xlim=(-3, 103),
                    ylim=(0.9, 3.25),
                )
                axis.text(
                    0.03,
                    0.96,
                    f"n = {len(q)}; main view: ratio"
                    f" 0.9–3.25\n{int(q.route_directness_ratio.gt(3.25).sum())} higher"
                    " outlier(s); full range below",
                    transform=axis.transAxes,
                    va="top",
                    fontsize=8,
                )
                inset = axis.inset_axes([0.08, 0.47, 0.38, 0.34])
                inset.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=5,
                    alpha=0.5,
                )
                inset.set(
                    xlim=(-3, 103),
                    ylim=(0.9, float(legs.route_directness_ratio.max()) + 0.5),
                    title="Full range (all legs)",
                )
                inset.title.set_fontsize(8)
                inset.tick_params(labelsize=6)
                axis.grid(alpha=0.15)
            axes[0].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            axes[1].tick_params(labelleft=True)
            value = {"within_region_associations": values}
    ax.grid(alpha=0.15)
    return json.loads(json.dumps(value, default=lambda x: x.item()))


evidence_value = evidence(23, 2, df, user_df, legs, visits, extra)
print(evidence_value)
plt.close("all")
plt.close("all")
# Display the verified visualization once.
display(Image(filename="visualizations/insight_2.png"))
```

### **Question 3**: Are more frequent road-class transitions associated with greater detour variability while speed variability can remain similar?

#### Analysis action

Interpret the constructed descriptive association, not an estimated causal effect.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

legs = pd.read_csv("movement_legs.csv")
visits = pd.read_csv("aoi_visits.csv")
extra = {}
extra["aois"] = pd.read_csv("aois.csv")


def evidence(case, index, df, user_df, legs, visits, extra):
    import json
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(8.4, 5.0), constrained_layout=True)
    moving = user_df[user_df.activity_state.eq("travelling")].copy()
    done = df[df.status.eq("completed")].copy()
    value = {}
    if case == 21:
        order = ["primary", "secondary", "tertiary", "residential"]
        moving = moving.merge(
            legs[["leg_id", "period"]], on="leg_id", validate="many_to_one"
        )
        if index == 1:
            summary = (
                moving.groupby("fclass")[["step_distance_km", "step_duration_s"]]
                .sum()
                .reindex(order)
            )
            summary["movement_speed_kmh"] = (
                summary.step_distance_km / summary.step_duration_s * 3600
            )
            summary["completions"] = done.delivery_fclass.value_counts().reindex(
                order, fill_value=0
            )
            summary["completions_per_km"] = (
                summary.completions / summary.step_distance_km
            )
            ax.bar(
                order,
                summary.movement_speed_kmh,
                color="#5c8ea7",
                alpha=0.8,
                label="Movement speed",
            )
            twin = ax.twinx()
            twin.plot(order, summary.completions_per_km, "o-", color="#bd7843")
            ax.set(ylabel="Distance / moving time (km/h)", xlabel="Native OSM fclass")
            twin.set_ylabel("Delivery completions per travelled km", color="#bd7843")
            value = {"road_class_summary": summary.to_dict("index")}
        elif index == 2:
            summary = (
                moving.groupby(["period", "fclass"])
                .step_distance_km.sum()
                .unstack(fill_value=0)
            )
            summary = summary.div(summary.sum(axis=1), axis=0).reindex(
                ["early", "late"]
            )
            summary[order].plot.bar(stacked=True, ax=ax, rot=0)
            ax.set(
                ylabel="Share of all moving distance",
                xlabel="First / second half of scheduled delivery stops",
            )
            tasks = done.merge(
                legs[["leg_id", "period"]], left_on="arrival_leg_id", right_on="leg_id"
            )
            value = {
                "distance_shares": summary.to_dict("index"),
                "completion_counts": tasks.period.value_counts().to_dict(),
            }
        else:
            # Smooth spatial AND temporal event kernel, independently measured
            # from task-completion positions/times rather than a generator label.
            frames = []
            for tid, q in moving.groupby("trajectory_id"):
                t = done[done.trajectory_id.eq(tid)]
                x = q[["lng", "lat"]].to_numpy() * [95.2, 111]
                y = t[["delivery_lng", "delivery_lat"]].to_numpy() * [95.2, 111]
                ds = np.linalg.norm(x[:, None] - y[None, :], axis=2)
                a = (
                    pd.to_datetime(q.gps_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                b = (
                    pd.to_datetime(t.completion_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                dt = abs(a[:, None] - b[None, :]) / 6e10
                q = q.copy()
                q["nearby_activity"] = (np.exp(-ds / 0.30) * np.exp(-dt / 30)).sum(
                    axis=1
                )
                frames.append(q)
            points = pd.concat(frames)
            values = {}
            for cls in order:
                q = points[points.fclass.eq(cls)]
                bins = pd.qcut(q.nearby_activity, 5, duplicates="drop")
                med = q.groupby(bins, observed=True)[
                    ["nearby_activity", "speed_kmh"]
                ].median()
                line = ax.plot(med.nearby_activity, med.speed_kmh, "o-", label=cls)[0]
                sample = q.sample(min(350, len(q)), random_state=17)
                ax.scatter(
                    sample.nearby_activity,
                    sample.speed_kmh,
                    s=6,
                    alpha=0.10,
                    color=line.get_color(),
                )
                values[cls] = {
                    "points": len(q),
                    "spearman": q.nearby_activity.corr(q.speed_kmh, method="spearman"),
                    "low_to_high_activity_speed_drop_kmh": (
                        med.speed_kmh.iloc[0] - med.speed_kmh.iloc[-1]
                    ),
                    "quintile_medians": med.to_dict("list"),
                }
            ax.set(
                xlabel=(
                    "Nearby completion intensity (300 m / 30 min exponential kernel)"
                ),
                ylabel="Median moving speed (km/h)",
            )
            ax.legend()
            value = {"within_class_evidence": values}
    elif case == 22:
        if index == 1:
            counts = (
                done.groupby(["service_date", "aoi_id"]).size().unstack(fill_value=0)
            )
            span = user_df.groupby("trajectory_id").gps_time.agg(["min", "max"])
            duration = (
                pd.to_datetime(span["max"], format="mixed")
                - pd.to_datetime(span["min"], format="mixed")
            ).dt.total_seconds()
            shares = (
                visits.groupby(["trajectory_id", "aoi_id"])
                .duration_s.sum()
                .unstack(fill_value=0)
                .div(duration, axis=0)
            )
            summary = pd.DataFrame(
                {
                    "mean_deliveries_per_day": counts.mean(),
                    "mean_work_time_share": shares.mean(),
                }
            )
            summary.plot.bar(
                y="mean_deliveries_per_day", ax=ax, rot=0, legend=False, color="#608ea4"
            )
            ax.set(
                ylabel="Completed deliveries per observed day", xlabel="Constructed AOI"
            )
            twin = ax.twinx()
            twin.plot(
                np.arange(len(summary)),
                summary.mean_work_time_share * 100,
                "o-",
                color="#bd7843",
            )
            twin.set_ylabel("Mean observed workday time (%)")
            value = {
                "aoi_summary": summary.to_dict("index"),
                "observed_days": counts.shape[0],
                "regular_destination_fclasses": sorted(
                    done[done.regular_aoi.eq("yes")].delivery_fclass.unique().tolist()
                ),
                "fixed_receipt_sites": (
                    extra["receipt_events"]
                    .groupby("aoi_id")[["lng", "lat"]]
                    .nunique()
                    .to_dict("index")
                ),
            }
        elif index == 2:
            q = legs[legs.purpose.eq("delivery")]
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.4), constrained_layout=True)
            ax = axes[0]
            ax.boxplot(
                [q[q.period.eq(p)].route_directness_ratio for p in ["early", "late"]],
                tick_labels=["Early", "Late"],
            )
            ax.set(
                xlabel="Delivery route phase",
                ylabel="Travelled / straight-line distance",
                title=(
                    "Later delivery routes are more direct\nLower ratio = less detour"
                ),
            )
            dates = sorted(q.service_date.unique())
            totals = moving.groupby("service_date")[
                ["step_distance_km", "step_duration_s"]
            ].sum()
            baseline = totals.step_distance_km / totals.step_duration_s * 3600
            late = q[q.period.eq("late")].copy()
            late["day_mean_movement_speed_kmh"] = late.service_date.map(baseline)
            late["difference_from_day_mean_kmh"] = (
                late.mean_speed_kmh - late.day_mean_movement_speed_kmh
            )
            above = late.difference_from_day_mean_kmh.gt(0)
            x = late.service_date.map({day: i for i, day in enumerate(dates)}).to_numpy(
                dtype=float
            )
            x += np.where(late.groupby("service_date").cumcount().eq(0), -0.12, 0.12)
            axes[1].axhline(
                0,
                color="#555555",
                linewidth=1,
                linestyle="--",
                label="Same as day mean",
            )
            for mask, color, label in [
                (above, "#3976a8", "Above day mean"),
                (~above, "#c77c3d", "At or below day mean"),
            ]:
                axes[1].scatter(
                    x[mask],
                    late.loc[mask, "difference_from_day_mean_kmh"],
                    color=color,
                    s=48,
                    label=f"{label}: {int(mask.sum())}/{len(late)} ({mask.mean():.1%})",
                )
            axes[1].set(
                xticks=np.arange(len(dates)),
                xticklabels=[day[5:] for day in dates],
                xlabel="Workday — each point is a delivery leg as time drew to a close",
                ylabel="Leg speed − day mean movement speed (km/h)",
                title=(
                    "As time drew to a close:\nMore direct does not guarantee"
                    " above-average speed"
                ),
            )
            axes[1].margins(y=0.30)
            axes[1].tick_params(axis="x", rotation=45, labelsize=8)
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "period_means": (
                    q.groupby("period")[["route_directness_ratio", "mean_speed_kmh"]]
                    .mean()
                    .to_dict("index")
                ),
                "day_mean_definition": (
                    "All moving GPS distance / all moving GPS time, including return"
                    " legs and excluding stops"
                ),
                "daily_mean_movement_speed_kmh": baseline.to_dict(),
                "late_delivery_legs": len(late),
                "late_legs_above_day_mean": int(above.sum()),
                "late_share_above_day_mean": float(above.mean()),
                "late_leg_comparison": (
                    late[
                        [
                            "leg_id",
                            "service_date",
                            "mean_speed_kmh",
                            "day_mean_movement_speed_kmh",
                            "difference_from_day_mean_kmh",
                        ]
                    ].to_dict("records")
                ),
            }
        elif index == 3:
            q = done.copy()
            stamp = pd.to_datetime(q.completion_time, format="mixed")
            q["completion_hour"] = (
                stamp.dt.hour + stamp.dt.minute / 60 + stamp.dt.second / 3600
            )
            q["window_hours"] = (
                pd.to_datetime(q.time_window_end) - pd.to_datetime(q.time_window_start)
            ).dt.total_seconds() / 3600
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13.5, 5.5), constrained_layout=True)
            ax = axes[0]
            groups = [
                q[q.regular_aoi.eq("no") & q.window_hours.lt(9)],
                q[q.regular_aoi.eq("no") & q.window_hours.ge(9)],
                q[q.regular_aoi.eq("yes") & q.window_hours.ge(9)],
            ]
            labels = [
                "Non-regular AOIs\nWindow <9 h",
                "Non-regular AOIs\nWindow ≥9 h",
                "Regular AOIs (A/B)\nWindow ≥9 h",
            ]
            colors = ["#4f846c", "#3976a8", "#c77c3d"]
            boxes = ax.boxplot(
                [g.completion_hour for g in groups],
                tick_labels=labels,
                patch_artist=True,
            )
            for box, color in zip(boxes["boxes"], colors):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, g in enumerate(groups, 1):
                median = g.completion_hour.median()
                h = int(median)
                minute = int(round((median - h) * 60))
                ax.text(
                    i,
                    17.8,
                    f"n={len(g)}\nMedian {h:02}:{minute:02}",
                    ha="center",
                    fontsize=9,
                )
            ax.set(
                ylabel="Task completion time (Shanghai)",
                title="Both non-regular window groups finish earlier",
                ylim=(7.5, 18.5),
            )
            ax.yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            ax.tick_params(axis="x", labelsize=8)
            for g, label, color, marker in zip(groups, labels, colors, ["o", "s", "^"]):
                axes[1].scatter(
                    g.window_hours,
                    g.completion_hour,
                    s=16,
                    alpha=0.45,
                    color=color,
                    marker=marker,
                    label=label.replace("\n", "; "),
                )
            axes[1].set(
                xlabel="Service-window duration (hours)",
                ylabel="Task completion time (Shanghai)",
                title="Wide windows alone do not explain later completion",
                ylim=(7.5, 19.5),
            )
            axes[1].yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(alpha=0.15)
            nvisits = visits.groupby(["service_date", "aoi_id"]).size()
            value = {
                "group_medians": (
                    q.groupby("regular_aoi")[["completion_hour", "window_hours"]]
                    .median()
                    .to_dict("index")
                ),
                "regular_aoi_visits_per_day": (
                    nvisits.unstack(fill_value=0)[["A", "B"]].to_dict("list")
                ),
                "irregular_by_window_median_hour": (
                    q[q.regular_aoi.eq("no")]
                    .assign(
                        window_group=lambda d: np.where(
                            d.window_hours < 9, "Narrower (<9h)", "Wide (9h+)"
                        )
                    )
                    .groupby("window_group")
                    .completion_hour.median()
                    .to_dict()
                ),
                "plotted_groups": {
                    label.replace("\n", "; "): {
                        "tasks": len(g),
                        "median_completion_hour": float(g.completion_hour.median()),
                        "window_hours_min": float(g.window_hours.min()),
                        "window_hours_max": float(g.window_hours.max()),
                    }
                    for label, g in zip(labels, groups)
                },
                "uncompleted_tasks_excluded": int(df.status.ne("completed").sum()),
            }
        else:
            irregular = df[df.regular_aoi.eq("no")]
            summary = (
                irregular.groupby(["service_date", "status"])
                .size()
                .unstack(fill_value=0)
            )
            summary.plot.bar(stacked=True, ax=ax, rot=50)
            ax.set(ylabel="Available irregular-AOI tasks", xlabel="Workday")
            pending = irregular[irregular.status.eq("pending")]
            dates = pending.service_date.unique().tolist()
            min_dist = {}
            for day in dates:
                gps = user_df[user_df.service_date.eq(day)]
                destinations = pending[pending.service_date.eq(day)][
                    ["aoi_id", "delivery_lng", "delivery_lat"]
                ].drop_duplicates()
                for r in destinations.itertuples():
                    ds = np.linalg.norm(
                        (
                            gps[["lng", "lat"]].to_numpy()
                            - [r.delivery_lng, r.delivery_lat]
                        )
                        * [95200, 111000],
                        axis=1,
                    )
                    min_dist[r.aoi_id] = float(ds.min())
            value = {
                "pending_dates": dates,
                "pending_tasks": len(pending),
                "minimum_GPS_distance_to_unvisited_AOI_m": min_dist,
                "anomaly_day_irregular_visit_count": int(
                    len(
                        visits[
                            visits.service_date.isin(dates)
                            & ~visits.aoi_id.isin(["A", "B"])
                        ]
                    )
                ),
            }
    elif case == 23:
        q = legs[legs.purpose.eq("delivery")].copy()
        # Initial departure is not between two completed task locations.
        q = q[q.groupby("trajectory_id").cumcount().gt(0)]
        if index == 1:
            q = q[q.period.ne("exception") & q.separation_km.between(0.8, 1.3)]
            q["oneway_group"] = np.where(
                q.oneway_share >= 0.55,
                "High (55%+)",
                np.where(q.oneway_share <= 0.25, "Low (0–25%)", "Middle"),
            )
            q = q[q.oneway_group.ne("Middle")]
            summary = q.groupby("oneway_group")[
                ["separation_km", "route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            high, low = summary.loc["High (55%+)"], summary.loc["Low (0–25%)"]
            relative = pd.Series(
                {
                    "Route directness": (
                        (high.route_directness_ratio / low.route_directness_ratio - 1)
                        * 100
                    ),
                    "Average speed": (
                        (high.mean_speed_kmh / low.mean_speed_kmh - 1) * 100
                    ),
                }
            )
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), constrained_layout=True)
            ax = axes[0]
            for name, s in q.groupby("oneway_group"):
                ax.scatter(
                    s.separation_km, s.route_directness_ratio, label=name, alpha=0.6
                )
            ax.set(
                xlabel="Task endpoint separation (km)",
                ylabel="Travelled / straight-line distance",
                title="Matched endpoint separations (0.8–1.3 km)",
            )
            ax.legend()
            bars = axes[1].bar(
                relative.index,
                relative.values,
                color=["#3976a8", "#c77c3d"],
                alpha=0.82,
            )
            axes[1].axhline(0, color="#666666", linewidth=1)
            axes[1].bar_label(
                bars,
                labels=[
                    (
                        f"+{relative.iloc[0]:.1f}%\n{low.route_directness_ratio:.2f} →"
                        f" {high.route_directness_ratio:.2f}"
                    ),
                    (
                        f"+{relative.iloc[1]:.1f}%\n{low.mean_speed_kmh:.1f} →"
                        f" {high.mean_speed_kmh:.1f} km/h"
                    ),
                ],
                padding=5,
                fontsize=10,
            )
            axes[1].set(
                ylabel="High one-way group relative to low group (%)",
                title="Average speed differs less than route directness",
                ylim=(0, max(relative) * 1.42),
            )
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "matched_separation_range_km": [0.8, 1.3],
                "group_means": summary.to_dict("index"),
                "high_vs_low_relative_difference_percent": relative.to_dict(),
                "within_period_group_means": (
                    q.groupby(["period", "oneway_group"])[
                        ["route_directness_ratio", "mean_speed_kmh"]
                    ]
                    .mean()
                    .reset_index()
                    .to_dict("records")
                ),
            }
        elif index == 2:
            means = (
                q[q.period.ne("exception")]
                .groupby("period")[["mean_speed_kmh", "route_directness_ratio"]]
                .mean()
                .reindex(["early", "midday", "late"])
            )
            stamp = pd.to_datetime(q.departure_time, format="mixed")
            q["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.floor(q.departure_hour.min() * 6) / 6,
                np.ceil(q.departure_hour.max() * 6) / 6 + 1 / 12,
                1 / 6,
            )
            windows = []
            for center in centers:
                subset = q[q.departure_hour.between(center - 0.5, center + 0.5)]
                windows.append(
                    {
                        "center_hour": float(center),
                        "legs": len(subset),
                        **{
                            column: (
                                float(subset[column].median())
                                if len(subset) >= 8
                                else None
                            )
                            for column in ["mean_speed_kmh", "route_directness_ratio"]
                        },
                    }
                )
            trend = pd.DataFrame(windows)
            plt.close(fig)
            fig, axes = plt.subplots(
                2, 1, figsize=(9.5, 7), sharex=True, constrained_layout=True
            )
            fig.suptitle("Speed and detour over actual departure time", fontsize=15)
            for axis, column, color, title in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "#3976a8",
                    "Movement speed of each task-to-task leg",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "#c77c3d",
                    "Route detour of the same legs at the same times",
                ),
            ]:
                normal = q[q.period.ne("exception")]
                exception = q[q.period.eq("exception")]
                axis.scatter(
                    normal.departure_hour,
                    normal[column],
                    s=22,
                    color=color,
                    alpha=0.45,
                    label=(
                        f"Task-to-task legs ({len(normal)};"
                        f" {normal.service_date.nunique()} days)"
                    ),
                )
                axis.scatter(
                    exception.departure_hour,
                    exception[column],
                    s=110,
                    color="#8c405f",
                    marker="*",
                    label="Exceptional leg (insight 4)",
                    zorder=4,
                )
                axis.plot(
                    trend.center_hour,
                    trend[column],
                    color=color,
                    linewidth=2.5,
                    label="60-min rolling median (≥8 legs)",
                )
                axis.set_title(title, fontsize=11)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(x=0.03, y=0.30)
                axis.legend(loc="upper left", fontsize=8, frameon=False)
            axes[0].set_ylabel("Mean movement speed (km/h)")
            axes[1].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            ticks = np.arange(
                np.floor(q.departure_hour.min()), np.ceil(q.departure_hour.max()) + 1
            )
            axes[1].set(
                xticks=ticks,
                xticklabels=[f"{int(h):02}:00" for h in ticks],
                xlabel="Actual leg departure time (Shanghai local time; days overlaid)",
            )
            value = {
                "period_means": means.to_dict("index"),
                "period_means_note": (
                    "Legacy phase aggregates retained for compatibility; not used to"
                    " draw this chart."
                ),
                "plotted_task_to_task_legs": len(q),
                "exception_legs_included": int(q.period.eq("exception").sum()),
                "rolling_window_minutes": 60,
                "minimum_legs_per_window": 8,
                "clock_time_rolling_medians": [
                    window for window in windows if window["legs"] >= 8
                ],
                "omitted_window_centers_hour": [
                    window["center_hour"] for window in windows if window["legs"] < 8
                ],
            }
        elif index == 3:
            p = (
                q[q.period.ne("exception")]
                .groupby(["service_date", "period"])
                .agg(
                    detour_sd=("route_directness_ratio", "std"),
                    speed_sd=("mean_speed_kmh", "std"),
                    transitions=("fclass_transitions", "sum"),
                    distance=("distance_km", "sum"),
                )
            )
            p["transitions_per_km"] = p.transitions / p.distance
            cut = p.transitions_per_km.median()
            p["transition_group"] = np.where(
                p.transitions_per_km > cut, "Frequent", "Less frequent"
            )
            pearson = p.transitions_per_km.corr(p.detour_sd)
            spearman = p.transitions_per_km.corr(p.detour_sd, method="spearman")
            group_means = p.groupby("transition_group")[
                ["detour_sd", "speed_sd", "transitions_per_km"]
            ].mean()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12.5, 5.5), constrained_layout=True)
            ax = axes[0]
            ax.scatter(
                p.transitions_per_km,
                p.detour_sd,
                color="#3976a8",
                s=43,
                alpha=0.72,
                label=f"Day × route phase (n={len(p)})",
            )
            slope, intercept = np.polyfit(p.transitions_per_km, p.detour_sd, 1)
            line = np.linspace(
                p.transitions_per_km.min(), p.transitions_per_km.max(), 100
            )
            ax.plot(
                line,
                slope * line + intercept,
                "--",
                color="#244d69",
                linewidth=2,
                label="Descriptive linear fit",
            )
            ax.axvline(
                cut,
                color="#777777",
                linestyle=":",
                linewidth=1.3,
                label=f"Median split: {cut:.2f}/km",
            )
            ax.set(
                xlabel="Native fclass transitions per travelled km",
                ylabel="Within-period route-ratio SD",
                title=(
                    "Detour variability increases with road-class switching\nPearson"
                    f" r={pearson:.2f}; Spearman ρ={spearman:.2f}"
                ),
            )
            ax.legend(loc="upper left", fontsize=8, frameon=False)
            ax.grid(alpha=0.15)
            order = ["Less frequent", "Frequent"]
            arrays = [p.loc[p.transition_group.eq(name), "speed_sd"] for name in order]
            boxes = axes[1].boxplot(
                arrays, tick_labels=order, patch_artist=True, showfliers=True
            )
            for box, color in zip(boxes["boxes"], ["#789bb0", "#d09a58"]):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, name in enumerate(order, 1):
                mean = group_means.loc[name, "speed_sd"]
                axes[1].scatter(
                    i,
                    mean,
                    marker="D",
                    s=45,
                    color="black",
                    zorder=4,
                    label="Mean" if i == 1 else None,
                )
                axes[1].annotate(
                    f"Mean {mean:.2f}",
                    (i, mean),
                    xytext=(8, 5),
                    textcoords="offset points",
                    fontsize=9,
                )
            axes[1].set(
                xlabel="Transition-frequency group (median split)",
                ylabel="Across-leg speed SD (km/h)",
                title=(
                    "Speed variability can remain similar\nIndividual periods still"
                    " vary"
                ),
            )
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "correlation_transition_frequency_detour_sd": pearson,
                "spearman_transition_frequency_detour_sd": spearman,
                "group_means": (
                    p.groupby("transition_group")[
                        ["detour_sd", "speed_sd", "transitions_per_km"]
                    ]
                    .mean()
                    .to_dict("index")
                ),
                "period_count": len(p),
            }
        else:
            normal = q[q.period.ne("exception")]
            exception = q[q.period.eq("exception")]
            pure = normal[np.isclose(normal.oneway_share, 1, atol=1e-10)]
            other = normal[~normal.index.isin(pure.index)]
            pure_mean = float(pure.route_directness_ratio.mean())
            ax.scatter(
                other.oneway_share * 100,
                other.route_directness_ratio,
                alpha=0.35,
                label="Other task legs",
            )
            ax.scatter(
                pure.oneway_share * 100,
                pure.route_directness_ratio,
                marker="D",
                s=55,
                color="#3976a8",
                label=f"Pure one-way routes (mean ratio {pure_mean:.2f})",
            )
            ax.scatter(
                exception.oneway_share * 100,
                exception.route_directness_ratio,
                marker="*",
                s=220,
                color="#c66643",
                label="Exceptional period",
            )
            ax.axhline(
                pure_mean, color="#3976a8", linestyle="--", linewidth=1.3, alpha=0.8
            )
            ax.set(
                xlabel="One-way-road distance (%)",
                ylabel="Travelled / straight-line distance",
            )
            ax.legend()
            value = {
                "exception": (
                    exception[
                        [
                            "leg_id",
                            "route_directness_ratio",
                            "oneway_share",
                            "mean_speed_kmh",
                            "distance_km",
                            "separation_km",
                        ]
                    ].to_dict("records")
                ),
                "normal_speed_10_90_percentiles": (
                    normal.mean_speed_kmh.quantile([0.1, 0.9]).tolist()
                ),
                "pure_oneway_reference": {
                    "definition": (
                        "delivery legs whose travelled distance is entirely on native"
                        " OSM one-way edges"
                    ),
                    "legs": int(len(pure)),
                    "mean_route_directness_ratio": pure_mean,
                    "route_directness_ratio_std": float(
                        pure.route_directness_ratio.std(ddof=0)
                    ),
                    "route_directness_ratios": pure.route_directness_ratio.tolist(),
                },
            }
    elif case == 24:
        if index == 1:
            q = legs[legs.purpose.str.startswith("road_condition")]
            q.boxplot(column="speed_std_kmh", by="purpose", ax=ax)
            fig.suptitle("")
            ax.set(
                xlabel="Road condition",
                ylabel="Short-term GPS speed SD (km/h)",
                title="Stable residential travel versus alternating road classes",
            )
            value = {
                "condition_means": (
                    q.groupby("purpose")[
                        [
                            "speed_std_kmh",
                            "residential_share",
                            "tertiary_share",
                            "fclass_transitions",
                        ]
                    ]
                    .mean()
                    .to_dict("index")
                )
            }
        else:
            q = extra["crossings"].merge(
                legs,
                left_on="crossing_leg_id",
                right_on="leg_id",
                validate="one_to_one",
            )
            low = moving[moving.speed_kmh < 12].groupby("leg_id").step_duration_s.sum()
            q["low_speed_seconds"] = q.leg_id.map(low).fillna(0)
            approach = legs.set_index("leg_id").mean_speed_kmh
            q["speed_drop_kmh"] = q.approach_leg_id.map(approach) - q.mean_speed_kmh
            if index == 2:
                # Fluctuation is within-crossing GPS speed SD, not variation
                # between the average speeds of different crossings.
                recomputed = moving.groupby("leg_id").speed_kmh.agg(
                    lambda s: float(np.std(s.to_numpy(), ddof=0))
                )
                np.testing.assert_allclose(
                    q.speed_std_kmh, q.leg_id.map(recomputed), atol=1e-8
                )
                plt.close(fig)
                fig, axes = plt.subplots(
                    1, 2, figsize=(12.5, 5.5), constrained_layout=True
                )
                ax = axes[0]
                structures = (
                    q[["structure_osm_way_id", "structure_type"]]
                    .drop_duplicates()
                    .sort_values("structure_type")
                )
                ticks = []
                labels = []
                for axis, column, title, ylabel in [
                    (
                        axes[0],
                        "low_speed_seconds",
                        "Longer low-speed time later in the workday",
                        "Time below 12 km/h per crossing (seconds)",
                    ),
                    (
                        axes[1],
                        "speed_std_kmh",
                        "Greater speed fluctuations later in the workday",
                        "Within-crossing GPS speed SD (km/h)",
                    ),
                ]:
                    for i, structure in enumerate(structures.itertuples()):
                        group = q[
                            q.structure_osm_way_id.eq(structure.structure_osm_way_id)
                        ]
                        paired = group.pivot(
                            index=["service_date", "approach_fclass"],
                            columns="period",
                            values=column,
                        )
                        assert paired[["early", "late"]].notna().all().all()
                        for row in paired.itertuples():
                            axis.plot(
                                [i * 3, i * 3 + 1],
                                [row.early, row.late],
                                color="#aab4bc",
                                alpha=0.5,
                                linewidth=0.8,
                            )
                        for j, (period, color) in enumerate(
                            [("early", "#3976a8"), ("late", "#c77c3d")]
                        ):
                            values = group.loc[group.period.eq(period), column]
                            axis.scatter(
                                np.full(len(values), i * 3 + j),
                                values,
                                s=24,
                                color=color,
                                alpha=0.7,
                                label=(
                                    period.capitalize() + " crossing"
                                    if i == 0
                                    else None
                                ),
                                zorder=3,
                            )
                            mean = values.mean()
                            axis.plot(
                                [i * 3 + j - 0.18, i * 3 + j + 0.18],
                                [mean, mean],
                                color="black",
                                linewidth=2,
                                label="Group mean" if i == 0 and j == 0 else None,
                                zorder=4,
                            )
                            axis.annotate(
                                f"{mean:.2f}",
                                (i * 3 + j, values.max()),
                                xytext=(0, 10),
                                textcoords="offset points",
                                ha="center",
                                fontweight="bold",
                                fontsize=10,
                            )
                        if axis is axes[0]:
                            ticks.extend([i * 3, i * 3 + 1])
                            labels.extend(
                                [
                                    f"Early\n{structure.structure_type.title()}",
                                    f"Late\n{structure.structure_type.title()}",
                                ]
                            )
                    axis.set(
                        xticks=ticks,
                        xticklabels=labels,
                        ylabel=ylabel,
                        title=title,
                        xlabel=(
                            "Paired within the same day, structure and approach class"
                        ),
                    )
                    axis.set_ylim(
                        0,
                        (
                            group[column].max() * 1.5
                            if column == "speed_std_kmh"
                            else q[column].max() * 1.5
                        ),
                    )
                    axis.legend(loc="upper left", fontsize=8, frameon=False)
                    axis.grid(axis="y", alpha=0.15)
                value = {
                    "structure_period_means": (
                        q.groupby(["structure_osm_way_id", "structure_type", "period"])[
                            ["low_speed_seconds", "speed_std_kmh", "mean_speed_kmh"]
                        ]
                        .mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                    "crossings": len(q),
                    "unique_structures": q.structure_osm_way_id.nunique(),
                }
            else:
                q.boxplot(column="speed_drop_kmh", by="approach_fclass", ax=ax)
                fig.suptitle("")
                ax.set(
                    xlabel="Dominant road class in last 800 m before structure",
                    ylabel="Approach minus crossing speed (km/h)",
                    title="Road approach conditions and crossing slowdowns",
                )
                value = {
                    "approach_means": (
                        q.groupby("approach_fclass")[
                            ["speed_drop_kmh", "approach_dominant_share"]
                        ]
                        .mean()
                        .to_dict("index")
                    ),
                    "within_structure_period_means": (
                        q.groupby(["structure_osm_way_id", "period", "approach_fclass"])
                        .speed_drop_kmh.mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                }
    elif case == 25:
        if index == 1:
            summary = legs.groupby("region_id")[
                ["route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            summary["delivery_volume"] = done.groupby("region_id").size()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12, 5.6), constrained_layout=True)
            ax = axes[0]
            order = ["Puxi_Jingan_West", "Puxi_Huangpu_East"]
            labels = [
                name + f'\n{int(summary.loc[name,"delivery_volume"])} deliveries'
                for name in order
            ]
            for axis, column, title, ylabel in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "Lower movement speed in Puxi_Jingan_West",
                    "Leg mean movement speed (km/h)",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "Smaller detour in Puxi_Jingan_West",
                    "Travelled / straight-line distance\nHigher = more detour",
                ),
            ]:
                arrays = [
                    legs.loc[legs.region_id.eq(name), column].to_numpy()
                    for name in order
                ]
                boxes = axis.boxplot(
                    arrays, tick_labels=labels, patch_artist=True, showfliers=True
                )
                for box, color in zip(boxes["boxes"], ["#3976a8", "#c77c3d"]):
                    box.set_facecolor(color)
                    box.set_alpha(0.35)
                for i, name in enumerate(order, 1):
                    mean = summary.loc[name, column]
                    axis.scatter(
                        i,
                        mean,
                        marker="D",
                        s=48,
                        color="black",
                        zorder=4,
                        label="Mean" if i == 1 else None,
                    )
                    axis.annotate(
                        f"Mean {mean:.2f}",
                        (i, mean),
                        xytext=(14, 5),
                        textcoords="offset points",
                        fontsize=9,
                        bbox=dict(
                            facecolor="white", alpha=0.8, edgecolor="none", pad=1
                        ),
                    )
                axis.set(title=title, ylabel=ylabel)
                axis.legend(loc="upper left", frameon=False, fontsize=9)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(y=0.18)
            difference = (
                summary.loc[order[1], "route_directness_ratio"]
                / summary.loc[order[0], "route_directness_ratio"]
                - 1
            )
            axes[1].set_title(
                f"Puxi_Huangpu_East: {difference:.1%} higher mean ratio", fontsize=11
            )
            axes[1].set_ylim(0.9, 3.25)
            outside = int(legs.route_directness_ratio.gt(3.25).sum())
            axes[1].text(
                0.03,
                0.88,
                f"Main view: 0.9–3.25\n{outside} higher outlier shown in inset",
                transform=axes[1].transAxes,
                fontsize=8,
                va="top",
            )
            inset = axes[1].inset_axes([0.07, 0.43, 0.37, 0.34])
            inset.boxplot(
                [
                    legs.loc[legs.region_id.eq(name), "route_directness_ratio"]
                    for name in order
                ],
                tick_labels=["Jingan_West", "Huangpu_East"],
                showfliers=True,
            )
            inset.set_title("Full range (all legs)", fontsize=8)
            inset.tick_params(labelsize=6)
            inset.grid(axis="y", alpha=0.15)
            value = {
                "region_means": summary.to_dict("index"),
                "region_role_mapping": (
                    extra["regions"][["region_id", "comparison_role"]].to_dict(
                        "records"
                    )
                ),
            }
        elif index == 2:
            summary = legs.groupby(["region_id", "period"])[
                ["route_directness_ratio", "primary_share", "residential_share"]
            ].mean()
            timed = legs.copy()
            stamp = pd.to_datetime(timed.departure_time, format="mixed")
            timed["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.ceil(timed.departure_hour.min() * 4) / 4,
                np.floor(timed.departure_hour.max() * 4) / 4 + 0.01,
                0.25,
            )
            curves = {}
            plt.close(fig)
            fig, ax = plt.subplots(figsize=(10, 6), constrained_layout=True)
            inset = ax.inset_axes([0.08, 0.48, 0.34, 0.26])
            for name, color in [
                ("Puxi_Jingan_West", "#3976a8"),
                ("Puxi_Huangpu_East", "#c77c3d"),
            ]:
                group = timed[timed.region_id.eq(name)]
                ax.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=15,
                    color=color,
                    alpha=0.25,
                )
                inset.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=5,
                    color=color,
                    alpha=0.4,
                )
                y = []
                records = []
                for center in centers:
                    window = group[
                        group.departure_hour.between(center - 0.5, center + 0.5)
                    ]
                    valid = len(window) >= 8 and window.service_date.nunique() >= 3
                    y.append(
                        float(window.route_directness_ratio.median())
                        if valid
                        else np.nan
                    )
                    if valid:
                        records.append(
                            dict(
                                center_hour=float(center),
                                median_route_ratio=y[-1],
                                legs=len(window),
                                days=int(window.service_date.nunique()),
                            )
                        )
                curves[name] = records
                ax.plot(
                    centers,
                    y,
                    "o-",
                    color=color,
                    linewidth=2.3,
                    markersize=4,
                    label=name,
                )
                inset.plot(centers, y, color=color, linewidth=1)
            ticks = np.arange(8, np.ceil(timed.departure_hour.max()) + 0.1, 0.5)
            labels = [f"{int(h):02}:{int(round((h%1)*60)):02}" for h in ticks]
            ax.set(
                xticks=ticks,
                xticklabels=labels,
                ylabel="Travelled / straight-line distance (higher = more detour)",
                xlabel="Actual departure time (Shanghai local time; 14 days overlaid)",
                title=(
                    "Regional detour trends over actual departure time\nDots:"
                    " individual legs; lines: 60-min rolling medians, every 15 min"
                ),
                ylim=(0.9, 3.25),
                xlim=(7.9, float(timed.departure_hour.max()) + 0.1),
            )
            inset.set(
                title="Full range (all legs)",
                xlim=ax.get_xlim(),
                ylim=(0.9, float(timed.route_directness_ratio.max()) + 0.5),
            )
            inset.title.set_fontsize(8)
            inset.tick_params(labelsize=7)
            inset.set_xticks([8, 10, 12], ["08:00", "10:00", "12:00"])
            ax.text(
                0.03,
                0.85,
                "Main view: ratio 0.9–3.25;"
                f" {int(timed.route_directness_ratio.gt(3.25).sum())} higher outlier in"
                " inset\nTrend requires ≥8 legs from ≥3 days per window; no"
                " interpolation",
                transform=ax.transAxes,
                fontsize=8,
                va="top",
            )
            ax.legend(loc="upper left", frameon=False)
            value = {
                "region_period_means": summary.reset_index().to_dict("records"),
                "period_means_note": (
                    "Legacy phase summary; plot uses actual departure times, not phase"
                    " labels."
                ),
                "rolling_window_minutes": 60,
                "evaluation_interval_minutes": 15,
                "minimum_legs": 8,
                "minimum_days": 3,
                "regional_clock_time_trends": curves,
            }
        else:
            values = {}
            plt.close(fig)
            fig, axes = plt.subplots(
                1,
                2,
                figsize=(12, 5.8),
                sharex=True,
                sharey=True,
                constrained_layout=True,
            )
            ax = axes[0]
            for axis, name, color in zip(
                axes, ["Puxi_Jingan_West", "Puxi_Huangpu_East"], ["#3976a8", "#c77c3d"]
            ):
                q = legs[legs.region_id.eq(name)]
                axis.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=24,
                    alpha=0.5,
                )
                values[name] = {
                    "pearson": q.oneway_share.corr(q.route_directness_ratio),
                    "spearman": q.oneway_share.corr(
                        q.route_directness_ratio, method="spearman"
                    ),
                    "legs": len(q),
                    "oneway_share_unique_values": q.oneway_share.nunique(),
                    "route_ratio_unique_values": q.route_directness_ratio.nunique(),
                }
                axis.set(
                    title=(
                        f'{name}\nPearson r = {values[name]["pearson"]:.2f}; Spearman ρ'
                        f' = {values[name]["spearman"]:.2f}'
                    ),
                    xlabel="One-way-road share of travelled distance (%)",
                    xlim=(-3, 103),
                    ylim=(0.9, 3.25),
                )
                axis.text(
                    0.03,
                    0.96,
                    f"n = {len(q)}; main view: ratio"
                    f" 0.9–3.25\n{int(q.route_directness_ratio.gt(3.25).sum())} higher"
                    " outlier(s); full range below",
                    transform=axis.transAxes,
                    va="top",
                    fontsize=8,
                )
                inset = axis.inset_axes([0.08, 0.47, 0.38, 0.34])
                inset.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=5,
                    alpha=0.5,
                )
                inset.set(
                    xlim=(-3, 103),
                    ylim=(0.9, float(legs.route_directness_ratio.max()) + 0.5),
                    title="Full range (all legs)",
                )
                inset.title.set_fontsize(8)
                inset.tick_params(labelsize=6)
                axis.grid(alpha=0.15)
            axes[0].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            axes[1].tick_params(labelleft=True)
            value = {"within_region_associations": values}
    ax.grid(alpha=0.15)
    return json.loads(json.dumps(value, default=lambda x: x.item()))


evidence_value = evidence(23, 3, df, user_df, legs, visits, extra)
print(evidence_value)
plt.close("all")
plt.close("all")
# Display the verified visualization once.
display(Image(filename="visualizations/insight_3.png"))
```

```python
{
  "data_type": "road_backed_evidence_3",
  "insight_type": "factor_relationship",
  "insight": "Periods with frequent transitions between different fclass categories show greater fluctuations in detour ratio, while movement speed can remain relatively stable.",
  "insight_value": {
    "correlation_transition_frequency_detour_sd": 0.6294253411348822,
    "spearman_transition_frequency_detour_sd": 0.5982142857142858,
    "group_means": {
      "Frequent": {
        "detour_sd": 0.22185905807378672,
        "speed_sd": 1.260469646921314,
        "transitions_per_km": 2.8706736778627726
      },
      "Less frequent": {
        "detour_sd": 0.1190576499019611,
        "speed_sd": 1.2233089662154195,
        "transitions_per_km": 1.3708971923585203
      }
    },
    "period_count": 63
  },
  "question": "Are more frequent road-class transitions associated with greater detour variability while speed variability can remain similar?",
  "actionable_insight": "Interpret the constructed descriptive association, not an estimated causal effect.",
  "plot": {
    "plot_type": "transition_detour_association_and_speed_variability",
    "title": "LaDe 23 insight 3",
    "file": "visualizations/insight_3.png",
    "description": "Each left-panel point is one service date × ordered route phase, with fclass transitions normalized by travelled distance and detour variability measured as the across-leg route-ratio SD. The dashed line is a descriptive linear fit; Pearson and Spearman correlations are labelled. The right panel uses the median transition-rate split and shows the full distributions of across-leg speed SD, not a claim that every frequent-transition period has stable speed. Diamonds show group means. The constructed association is descriptive, not causal."
  },
  "code": "import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nlegs = pd.read_csv(\"movement_legs.csv\")\nvisits = pd.read_csv(\"aoi_visits.csv\")\nextra = {}\nextra[\"aois\"] = pd.read_csv(\"aois.csv\")\n\ndef evidence(case, index, df, user_df, legs, visits, extra):\n    import json\n    import numpy as np\n    import pandas as pd\n    import matplotlib.pyplot as plt\n\n    fig, ax = plt.subplots(figsize=(8.4, 5.0), constrained_layout=True)\n    moving = user_df[user_df.activity_state.eq('travelling')].copy()\n    done = df[df.status.eq('completed')].copy()\n    value = {}\n    if case == 21:\n        order = ['primary', 'secondary', 'tertiary', 'residential']\n        moving = moving.merge(legs[['leg_id', 'period']], on='leg_id', validate='many_to_one')\n        if index == 1:\n            summary = moving.groupby('fclass')[['step_distance_km', 'step_duration_s']].sum().reindex(order)\n            summary['movement_speed_kmh'] = summary.step_distance_km/summary.step_duration_s*3600\n            summary['completions'] = done.delivery_fclass.value_counts().reindex(order, fill_value=0)\n            summary['completions_per_km'] = summary.completions/summary.step_distance_km\n            ax.bar(order, summary.movement_speed_kmh, color='#5c8ea7', alpha=.8, label='Movement speed')\n            twin = ax.twinx(); twin.plot(order, summary.completions_per_km, 'o-', color='#bd7843')\n            ax.set(ylabel='Distance / moving time (km/h)', xlabel='Native OSM fclass')\n            twin.set_ylabel('Delivery completions per travelled km', color='#bd7843')\n            value = {'road_class_summary': summary.to_dict('index')}\n        elif index == 2:\n            summary = moving.groupby(['period', 'fclass']).step_distance_km.sum().unstack(fill_value=0)\n            summary = summary.div(summary.sum(axis=1), axis=0).reindex(['early', 'late'])\n            summary[order].plot.bar(stacked=True, ax=ax, rot=0)\n            ax.set(ylabel='Share of all moving distance', xlabel='First / second half of scheduled delivery stops')\n            tasks = done.merge(legs[['leg_id', 'period']], left_on='arrival_leg_id', right_on='leg_id')\n            value = {'distance_shares': summary.to_dict('index'), 'completion_counts': tasks.period.value_counts().to_dict()}\n        else:\n            # Smooth spatial AND temporal event kernel, independently measured\n            # from task-completion positions/times rather than a generator label.\n            frames = []\n            for tid, q in moving.groupby('trajectory_id'):\n                t = done[done.trajectory_id.eq(tid)]\n                x = q[['lng', 'lat']].to_numpy()*[95.2, 111]\n                y = t[['delivery_lng', 'delivery_lat']].to_numpy()*[95.2, 111]\n                ds = np.linalg.norm(x[:, None]-y[None, :], axis=2)\n                a = pd.to_datetime(q.gps_time, format='mixed').astype('int64').to_numpy()\n                b = pd.to_datetime(t.completion_time, format='mixed').astype('int64').to_numpy()\n                dt = abs(a[:, None]-b[None, :])/6e10\n                q = q.copy(); q['nearby_activity'] = (np.exp(-ds/.30)*np.exp(-dt/30)).sum(axis=1)\n                frames.append(q)\n            points = pd.concat(frames); values = {}\n            for cls in order:\n                q = points[points.fclass.eq(cls)]\n                bins = pd.qcut(q.nearby_activity, 5, duplicates='drop')\n                med = q.groupby(bins, observed=True)[['nearby_activity', 'speed_kmh']].median()\n                line = ax.plot(med.nearby_activity, med.speed_kmh, 'o-', label=cls)[0]\n                sample = q.sample(min(350, len(q)), random_state=17)\n                ax.scatter(sample.nearby_activity, sample.speed_kmh, s=6, alpha=.10, color=line.get_color())\n                values[cls] = {'points': len(q), 'spearman': q.nearby_activity.corr(q.speed_kmh, method='spearman'),\n                    'low_to_high_activity_speed_drop_kmh': med.speed_kmh.iloc[0]-med.speed_kmh.iloc[-1],\n                    'quintile_medians': med.to_dict('list')}\n            ax.set(xlabel='Nearby completion intensity (300 m / 30 min exponential kernel)', ylabel='Median moving speed (km/h)'); ax.legend()\n            value = {'within_class_evidence': values}\n    elif case == 22:\n        if index == 1:\n            counts = done.groupby(['service_date', 'aoi_id']).size().unstack(fill_value=0)\n            span = user_df.groupby('trajectory_id').gps_time.agg(['min', 'max'])\n            duration = (pd.to_datetime(span['max'], format='mixed')-pd.to_datetime(span['min'], format='mixed')).dt.total_seconds()\n            shares = visits.groupby(['trajectory_id', 'aoi_id']).duration_s.sum().unstack(fill_value=0).div(duration, axis=0)\n            summary = pd.DataFrame({'mean_deliveries_per_day': counts.mean(), 'mean_work_time_share': shares.mean()})\n            summary.plot.bar(y='mean_deliveries_per_day', ax=ax, rot=0, legend=False, color='#608ea4')\n            ax.set(ylabel='Completed deliveries per observed day', xlabel='Constructed AOI')\n            twin = ax.twinx(); twin.plot(np.arange(len(summary)), summary.mean_work_time_share*100, 'o-', color='#bd7843'); twin.set_ylabel('Mean observed workday time (%)')\n            value = {'aoi_summary': summary.to_dict('index'), 'observed_days': counts.shape[0],\n                'regular_destination_fclasses': sorted(done[done.regular_aoi.eq('yes')].delivery_fclass.unique().tolist()),\n                'fixed_receipt_sites': extra['receipt_events'].groupby('aoi_id')[['lng', 'lat']].nunique().to_dict('index')}\n        elif index == 2:\n            q = legs[legs.purpose.eq('delivery')]\n            plt.close(fig)\n            fig,axes=plt.subplots(1,2,figsize=(13,5.4),constrained_layout=True);ax=axes[0]\n            ax.boxplot([q[q.period.eq(p)].route_directness_ratio for p in ['early','late']],\n                tick_labels=['Early','Late'])\n            ax.set(xlabel='Delivery route phase',ylabel='Travelled / straight-line distance',\n                title='Later delivery routes are more direct\\nLower ratio = less detour')\n            dates=sorted(q.service_date.unique())\n            totals=moving.groupby('service_date')[['step_distance_km','step_duration_s']].sum()\n            baseline=totals.step_distance_km/totals.step_duration_s*3600\n            late=q[q.period.eq('late')].copy()\n            late['day_mean_movement_speed_kmh']=late.service_date.map(baseline)\n            late['difference_from_day_mean_kmh']=late.mean_speed_kmh-late.day_mean_movement_speed_kmh\n            above=late.difference_from_day_mean_kmh.gt(0)\n            x=late.service_date.map({day:i for i,day in enumerate(dates)}).to_numpy(dtype=float)\n            x+=np.where(late.groupby('service_date').cumcount().eq(0),-.12,.12)\n            axes[1].axhline(0,color='#555555',linewidth=1,linestyle='--',label='Same as day mean')\n            for mask,color,label in [(above,'#3976a8','Above day mean'),(~above,'#c77c3d','At or below day mean')]:\n                axes[1].scatter(x[mask],late.loc[mask,'difference_from_day_mean_kmh'],color=color,s=48,\n                    label=f'{label}: {int(mask.sum())}/{len(late)} ({mask.mean():.1%})')\n            axes[1].set(xticks=np.arange(len(dates)),xticklabels=[day[5:] for day in dates],\n                xlabel='Workday — each point is a delivery leg as time drew to a close',ylabel='Leg speed − day mean movement speed (km/h)',\n                title='As time drew to a close:\\nMore direct does not guarantee above-average speed')\n            axes[1].margins(y=.30)\n            axes[1].tick_params(axis='x',rotation=45,labelsize=8)\n            axes[1].legend(loc='upper left',fontsize=8,frameon=False)\n            axes[1].grid(axis='y',alpha=.15)\n            value = {'period_means': q.groupby('period')[['route_directness_ratio', 'mean_speed_kmh']].mean().to_dict('index'),\n                'day_mean_definition':'All moving GPS distance / all moving GPS time, including return legs and excluding stops',\n                'daily_mean_movement_speed_kmh':baseline.to_dict(),\n                'late_delivery_legs':len(late),'late_legs_above_day_mean':int(above.sum()),\n                'late_share_above_day_mean':float(above.mean()),\n                'late_leg_comparison':late[['leg_id','service_date','mean_speed_kmh','day_mean_movement_speed_kmh','difference_from_day_mean_kmh']].to_dict('records')}\n        elif index == 3:\n            q = done.copy(); stamp = pd.to_datetime(q.completion_time, format='mixed')\n            q['completion_hour'] = stamp.dt.hour+stamp.dt.minute/60+stamp.dt.second/3600\n            q['window_hours'] = (pd.to_datetime(q.time_window_end)-pd.to_datetime(q.time_window_start)).dt.total_seconds()/3600\n            plt.close(fig)\n            fig,axes=plt.subplots(1,2,figsize=(13.5,5.5),constrained_layout=True);ax=axes[0]\n            groups=[q[q.regular_aoi.eq('no')&q.window_hours.lt(9)],\n                    q[q.regular_aoi.eq('no')&q.window_hours.ge(9)],q[q.regular_aoi.eq('yes')&q.window_hours.ge(9)]]\n            labels=['Non-regular AOIs\\nWindow <9 h','Non-regular AOIs\\nWindow ≥9 h','Regular AOIs (A/B)\\nWindow ≥9 h']\n            colors=['#4f846c','#3976a8','#c77c3d']\n            boxes=ax.boxplot([g.completion_hour for g in groups],tick_labels=labels,patch_artist=True)\n            for box,color in zip(boxes['boxes'],colors):box.set_facecolor(color);box.set_alpha(.45)\n            for i,g in enumerate(groups,1):\n                median=g.completion_hour.median();h=int(median);minute=int(round((median-h)*60))\n                ax.text(i,17.8,f'n={len(g)}\\nMedian {h:02}:{minute:02}',ha='center',fontsize=9)\n            ax.set(ylabel='Task completion time (Shanghai)',title='Both non-regular window groups finish earlier',ylim=(7.5,18.5))\n            ax.yaxis.set_major_formatter(lambda v,_:f'{int(v):02}:00')\n            ax.tick_params(axis='x',labelsize=8)\n            for g,label,color,marker in zip(groups,labels,colors,['o','s','^']):\n                axes[1].scatter(g.window_hours,g.completion_hour,s=16,alpha=.45,\n                    color=color,marker=marker,label=label.replace('\\n','; '))\n            axes[1].set(xlabel='Service-window duration (hours)',ylabel='Task completion time (Shanghai)',\n                title='Wide windows alone do not explain later completion',ylim=(7.5,19.5))\n            axes[1].yaxis.set_major_formatter(lambda v,_:f'{int(v):02}:00')\n            axes[1].legend(loc='upper left',fontsize=8,frameon=False)\n            axes[1].grid(alpha=.15)\n            nvisits = visits.groupby(['service_date', 'aoi_id']).size()\n            value = {'group_medians': q.groupby('regular_aoi')[['completion_hour', 'window_hours']].median().to_dict('index'),\n                'regular_aoi_visits_per_day': nvisits.unstack(fill_value=0)[['A', 'B']].to_dict('list'),\n                'irregular_by_window_median_hour': q[q.regular_aoi.eq('no')].assign(window_group=lambda d: np.where(d.window_hours < 9, 'Narrower (<9h)', 'Wide (9h+)')).groupby('window_group').completion_hour.median().to_dict(),\n                'plotted_groups':{label.replace('\\n','; '):{'tasks':len(g),'median_completion_hour':float(g.completion_hour.median()),\n                    'window_hours_min':float(g.window_hours.min()),'window_hours_max':float(g.window_hours.max())}\n                    for label,g in zip(labels,groups)},\n                'uncompleted_tasks_excluded':int(df.status.ne('completed').sum())}\n        else:\n            irregular = df[df.regular_aoi.eq('no')]\n            summary = irregular.groupby(['service_date', 'status']).size().unstack(fill_value=0)\n            summary.plot.bar(stacked=True, ax=ax, rot=50)\n            ax.set(ylabel='Available irregular-AOI tasks', xlabel='Workday')\n            pending = irregular[irregular.status.eq('pending')]; dates = pending.service_date.unique().tolist()\n            min_dist = {}\n            for day in dates:\n                gps = user_df[user_df.service_date.eq(day)]\n                destinations = pending[pending.service_date.eq(day)][['aoi_id', 'delivery_lng', 'delivery_lat']].drop_duplicates()\n                for r in destinations.itertuples():\n                    ds = np.linalg.norm((gps[['lng', 'lat']].to_numpy()-[r.delivery_lng, r.delivery_lat])*[95200, 111000], axis=1)\n                    min_dist[r.aoi_id] = float(ds.min())\n            value = {'pending_dates': dates, 'pending_tasks': len(pending), 'minimum_GPS_distance_to_unvisited_AOI_m': min_dist,\n                'anomaly_day_irregular_visit_count': int(len(visits[visits.service_date.isin(dates)&~visits.aoi_id.isin(['A', 'B'])]))}\n    elif case == 23:\n        q = legs[legs.purpose.eq('delivery')].copy()\n        # Initial departure is not between two completed task locations.\n        q = q[q.groupby('trajectory_id').cumcount().gt(0)]\n        if index == 1:\n            q = q[q.period.ne('exception')&q.separation_km.between(.8, 1.3)]\n            q['oneway_group'] = np.where(q.oneway_share >= .55, 'High (55%+)', np.where(q.oneway_share <= .25, 'Low (0–25%)', 'Middle'))\n            q = q[q.oneway_group.ne('Middle')]\n            summary = q.groupby('oneway_group')[['separation_km', 'route_directness_ratio', 'mean_speed_kmh']].mean()\n            high, low = summary.loc['High (55%+)'], summary.loc['Low (0–25%)']\n            relative = pd.Series({\n                'Route directness': (high.route_directness_ratio/low.route_directness_ratio-1)*100,\n                'Average speed': (high.mean_speed_kmh/low.mean_speed_kmh-1)*100})\n            plt.close(fig)\n            fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), constrained_layout=True); ax = axes[0]\n            for name, s in q.groupby('oneway_group'):\n                ax.scatter(s.separation_km, s.route_directness_ratio, label=name, alpha=.6)\n            ax.set(xlabel='Task endpoint separation (km)', ylabel='Travelled / straight-line distance',\n                title='Matched endpoint separations (0.8–1.3 km)'); ax.legend()\n            bars = axes[1].bar(relative.index, relative.values, color=['#3976a8', '#c77c3d'], alpha=.82)\n            axes[1].axhline(0, color='#666666', linewidth=1)\n            axes[1].bar_label(bars, labels=[\n                f'+{relative.iloc[0]:.1f}%\\n{low.route_directness_ratio:.2f} → {high.route_directness_ratio:.2f}',\n                f'+{relative.iloc[1]:.1f}%\\n{low.mean_speed_kmh:.1f} → {high.mean_speed_kmh:.1f} km/h'],\n                padding=5, fontsize=10)\n            axes[1].set(ylabel='High one-way group relative to low group (%)',\n                title='Average speed differs less than route directness',\n                ylim=(0, max(relative)*1.42))\n            axes[1].grid(axis='y', alpha=.15)\n            value = {'matched_separation_range_km': [.8, 1.3], 'group_means': summary.to_dict('index'),\n                'high_vs_low_relative_difference_percent': relative.to_dict(),\n                'within_period_group_means': q.groupby(['period', 'oneway_group'])[['route_directness_ratio', 'mean_speed_kmh']].mean().reset_index().to_dict('records')}\n        elif index == 2:\n            means = q[q.period.ne('exception')].groupby('period')[['mean_speed_kmh', 'route_directness_ratio']].mean().reindex(['early', 'midday', 'late'])\n            stamp=pd.to_datetime(q.departure_time,format='mixed')\n            q['departure_hour']=(stamp-stamp.dt.normalize()).dt.total_seconds()/3600\n            centers=np.arange(np.floor(q.departure_hour.min()*6)/6,np.ceil(q.departure_hour.max()*6)/6+1/12,1/6)\n            windows=[]\n            for center in centers:\n                subset=q[q.departure_hour.between(center-.5,center+.5)]\n                windows.append({'center_hour':float(center),'legs':len(subset),\n                    **{column:float(subset[column].median()) if len(subset)>=8 else None\n                       for column in ['mean_speed_kmh','route_directness_ratio']}})\n            trend=pd.DataFrame(windows)\n            plt.close(fig)\n            fig,axes=plt.subplots(2,1,figsize=(9.5,7),sharex=True,constrained_layout=True)\n            fig.suptitle('Speed and detour over actual departure time',fontsize=15)\n            for axis,column,color,title in [\n                (axes[0],'mean_speed_kmh','#3976a8','Movement speed of each task-to-task leg'),\n                (axes[1],'route_directness_ratio','#c77c3d','Route detour of the same legs at the same times')]:\n                normal=q[q.period.ne('exception')];exception=q[q.period.eq('exception')]\n                axis.scatter(normal.departure_hour,normal[column],s=22,color=color,alpha=.45,\n                    label=f'Task-to-task legs ({len(normal)}; {normal.service_date.nunique()} days)')\n                axis.scatter(exception.departure_hour,exception[column],s=110,color='#8c405f',marker='*',\n                    label='Exceptional leg (insight 4)',zorder=4)\n                axis.plot(trend.center_hour,trend[column],color=color,linewidth=2.5,\n                    label='60-min rolling median (≥8 legs)')\n                axis.set_title(title,fontsize=11)\n                axis.grid(axis='y',alpha=.15)\n                axis.margins(x=.03,y=.30)\n                axis.legend(loc='upper left',fontsize=8,frameon=False)\n            axes[0].set_ylabel('Mean movement speed (km/h)')\n            axes[1].set_ylabel('Travelled / straight-line distance\\nHigher = more detour')\n            ticks=np.arange(np.floor(q.departure_hour.min()),np.ceil(q.departure_hour.max())+1)\n            axes[1].set(xticks=ticks,xticklabels=[f'{int(h):02}:00' for h in ticks],\n                xlabel='Actual leg departure time (Shanghai local time; days overlaid)')\n            value = {'period_means': means.to_dict('index'),\n                'period_means_note':'Legacy phase aggregates retained for compatibility; not used to draw this chart.',\n                'plotted_task_to_task_legs':len(q),'exception_legs_included':int(q.period.eq('exception').sum()),\n                'rolling_window_minutes':60,'minimum_legs_per_window':8,\n                'clock_time_rolling_medians':[window for window in windows if window['legs']>=8],\n                'omitted_window_centers_hour':[window['center_hour'] for window in windows if window['legs']<8]}\n        elif index == 3:\n            p = q[q.period.ne('exception')].groupby(['service_date', 'period']).agg(detour_sd=('route_directness_ratio', 'std'), speed_sd=('mean_speed_kmh', 'std'), transitions=('fclass_transitions', 'sum'), distance=('distance_km', 'sum'))\n            p['transitions_per_km'] = p.transitions/p.distance\n            cut = p.transitions_per_km.median(); p['transition_group'] = np.where(p.transitions_per_km > cut, 'Frequent', 'Less frequent')\n            pearson=p.transitions_per_km.corr(p.detour_sd)\n            spearman=p.transitions_per_km.corr(p.detour_sd,method='spearman')\n            group_means=p.groupby('transition_group')[['detour_sd','speed_sd','transitions_per_km']].mean()\n            plt.close(fig)\n            fig,axes=plt.subplots(1,2,figsize=(12.5,5.5),constrained_layout=True);ax=axes[0]\n            ax.scatter(p.transitions_per_km,p.detour_sd,color='#3976a8',s=43,alpha=.72,\n                label=f'Day × route phase (n={len(p)})')\n            slope,intercept=np.polyfit(p.transitions_per_km,p.detour_sd,1)\n            line=np.linspace(p.transitions_per_km.min(),p.transitions_per_km.max(),100)\n            ax.plot(line,slope*line+intercept,'--',color='#244d69',linewidth=2,label='Descriptive linear fit')\n            ax.axvline(cut,color='#777777',linestyle=':',linewidth=1.3,label=f'Median split: {cut:.2f}/km')\n            ax.set(xlabel='Native fclass transitions per travelled km',ylabel='Within-period route-ratio SD',\n                title=f'Detour variability increases with road-class switching\\nPearson r={pearson:.2f}; Spearman ρ={spearman:.2f}')\n            ax.legend(loc='upper left',fontsize=8,frameon=False);ax.grid(alpha=.15)\n            order=['Less frequent','Frequent'];arrays=[p.loc[p.transition_group.eq(name),'speed_sd'] for name in order]\n            boxes=axes[1].boxplot(arrays,tick_labels=order,patch_artist=True,showfliers=True)\n            for box,color in zip(boxes['boxes'],['#789bb0','#d09a58']):box.set_facecolor(color);box.set_alpha(.45)\n            for i,name in enumerate(order,1):\n                mean=group_means.loc[name,'speed_sd']\n                axes[1].scatter(i,mean,marker='D',s=45,color='black',zorder=4,label='Mean' if i==1 else None)\n                axes[1].annotate(f'Mean {mean:.2f}',(i,mean),xytext=(8,5),textcoords='offset points',fontsize=9)\n            axes[1].set(xlabel='Transition-frequency group (median split)',ylabel='Across-leg speed SD (km/h)',\n                title='Speed variability can remain similar\\nIndividual periods still vary')\n            axes[1].legend(loc='upper left',fontsize=8,frameon=False);axes[1].grid(axis='y',alpha=.15)\n            value = {'correlation_transition_frequency_detour_sd': pearson,\n                'spearman_transition_frequency_detour_sd':spearman,\n                'group_means': p.groupby('transition_group')[['detour_sd', 'speed_sd', 'transitions_per_km']].mean().to_dict('index'), 'period_count': len(p)}\n        else:\n            normal = q[q.period.ne('exception')]; exception = q[q.period.eq('exception')]\n            pure = normal[np.isclose(normal.oneway_share, 1, atol=1e-10)]\n            other = normal[~normal.index.isin(pure.index)]\n            pure_mean = float(pure.route_directness_ratio.mean())\n            ax.scatter(other.oneway_share*100, other.route_directness_ratio, alpha=.35,\n                label='Other task legs')\n            ax.scatter(pure.oneway_share*100, pure.route_directness_ratio, marker='D', s=55,\n                color='#3976a8', label=f'Pure one-way routes (mean ratio {pure_mean:.2f})')\n            ax.scatter(exception.oneway_share*100, exception.route_directness_ratio, marker='*', s=220, color='#c66643', label='Exceptional period')\n            ax.axhline(pure_mean, color='#3976a8', linestyle='--', linewidth=1.3, alpha=.8)\n            ax.set(xlabel='One-way-road distance (%)', ylabel='Travelled / straight-line distance'); ax.legend()\n            value = {'exception': exception[['leg_id', 'route_directness_ratio', 'oneway_share', 'mean_speed_kmh', 'distance_km', 'separation_km']].to_dict('records'),\n                'normal_speed_10_90_percentiles': normal.mean_speed_kmh.quantile([.1, .9]).tolist(),\n                'pure_oneway_reference': {'definition': 'delivery legs whose travelled distance is entirely on native OSM one-way edges',\n                    'legs': int(len(pure)), 'mean_route_directness_ratio': pure_mean,\n                    'route_directness_ratio_std': float(pure.route_directness_ratio.std(ddof=0)),\n                    'route_directness_ratios': pure.route_directness_ratio.tolist()}}\n    elif case == 24:\n        if index == 1:\n            q = legs[legs.purpose.str.startswith('road_condition')]\n            q.boxplot(column='speed_std_kmh', by='purpose', ax=ax); fig.suptitle('')\n            ax.set(xlabel='Road condition', ylabel='Short-term GPS speed SD (km/h)', title='Stable residential travel versus alternating road classes')\n            value = {'condition_means': q.groupby('purpose')[['speed_std_kmh', 'residential_share', 'tertiary_share', 'fclass_transitions']].mean().to_dict('index')}\n        else:\n            q = extra['crossings'].merge(legs, left_on='crossing_leg_id', right_on='leg_id', validate='one_to_one')\n            low = moving[moving.speed_kmh < 12].groupby('leg_id').step_duration_s.sum()\n            q['low_speed_seconds'] = q.leg_id.map(low).fillna(0)\n            approach = legs.set_index('leg_id').mean_speed_kmh\n            q['speed_drop_kmh'] = q.approach_leg_id.map(approach)-q.mean_speed_kmh\n            if index == 2:\n                # Fluctuation is within-crossing GPS speed SD, not variation\n                # between the average speeds of different crossings.\n                recomputed=moving.groupby('leg_id').speed_kmh.agg(lambda s:float(np.std(s.to_numpy(),ddof=0)))\n                np.testing.assert_allclose(q.speed_std_kmh,q.leg_id.map(recomputed),atol=1e-8)\n                plt.close(fig)\n                fig,axes=plt.subplots(1,2,figsize=(12.5,5.5),constrained_layout=True);ax=axes[0]\n                structures=q[['structure_osm_way_id','structure_type']].drop_duplicates().sort_values('structure_type')\n                ticks=[];labels=[]\n                for axis,column,title,ylabel in [\n                    (axes[0],'low_speed_seconds','Longer low-speed time later in the workday','Time below 12 km/h per crossing (seconds)'),\n                    (axes[1],'speed_std_kmh','Greater speed fluctuations later in the workday','Within-crossing GPS speed SD (km/h)')]:\n                    for i,structure in enumerate(structures.itertuples()):\n                        group=q[q.structure_osm_way_id.eq(structure.structure_osm_way_id)]\n                        paired=group.pivot(index=['service_date','approach_fclass'],columns='period',values=column)\n                        assert paired[['early','late']].notna().all().all()\n                        for row in paired.itertuples():\n                            axis.plot([i*3,i*3+1],[row.early,row.late],color='#aab4bc',alpha=.5,linewidth=.8)\n                        for j,(period,color) in enumerate([('early','#3976a8'),('late','#c77c3d')]):\n                            values=group.loc[group.period.eq(period),column]\n                            axis.scatter(np.full(len(values),i*3+j),values,s=24,color=color,alpha=.7,\n                                label=period.capitalize()+' crossing' if i==0 else None,zorder=3)\n                            mean=values.mean()\n                            axis.plot([i*3+j-.18,i*3+j+.18],[mean,mean],color='black',linewidth=2,\n                                label='Group mean' if i==0 and j==0 else None,zorder=4)\n                            axis.annotate(f'{mean:.2f}',(i*3+j,values.max()),xytext=(0,10),textcoords='offset points',\n                                ha='center',fontweight='bold',fontsize=10)\n                        if axis is axes[0]:\n                            ticks.extend([i*3,i*3+1]);labels.extend([f'Early\\n{structure.structure_type.title()}',f'Late\\n{structure.structure_type.title()}'])\n                    axis.set(xticks=ticks,xticklabels=labels,ylabel=ylabel,title=title,\n                        xlabel='Paired within the same day, structure and approach class')\n                    axis.set_ylim(0,group[column].max()*1.5 if column=='speed_std_kmh' else q[column].max()*1.5)\n                    axis.legend(loc='upper left',fontsize=8,frameon=False)\n                    axis.grid(axis='y',alpha=.15)\n                value = {'structure_period_means': q.groupby(['structure_osm_way_id', 'structure_type', 'period'])[['low_speed_seconds', 'speed_std_kmh', 'mean_speed_kmh']].mean().reset_index().to_dict('records'),\n                    'crossings': len(q), 'unique_structures': q.structure_osm_way_id.nunique()}\n            else:\n                q.boxplot(column='speed_drop_kmh', by='approach_fclass', ax=ax); fig.suptitle('')\n                ax.set(xlabel='Dominant road class in last 800 m before structure', ylabel='Approach minus crossing speed (km/h)', title='Road approach conditions and crossing slowdowns')\n                value = {'approach_means': q.groupby('approach_fclass')[['speed_drop_kmh', 'approach_dominant_share']].mean().to_dict('index'),\n                    'within_structure_period_means': q.groupby(['structure_osm_way_id', 'period', 'approach_fclass']).speed_drop_kmh.mean().reset_index().to_dict('records')}\n    elif case == 25:\n        if index == 1:\n            summary = legs.groupby('region_id')[['route_directness_ratio', 'mean_speed_kmh']].mean()\n            summary['delivery_volume'] = done.groupby('region_id').size()\n            plt.close(fig)\n            fig,axes=plt.subplots(1,2,figsize=(12,5.6),constrained_layout=True);ax=axes[0]\n            order=['Puxi_Jingan_West','Puxi_Huangpu_East']\n            labels=[name+f'\\n{int(summary.loc[name,\"delivery_volume\"])} deliveries' for name in order]\n            for axis,column,title,ylabel in [\n                (axes[0],'mean_speed_kmh','Lower movement speed in Puxi_Jingan_West','Leg mean movement speed (km/h)'),\n                (axes[1],'route_directness_ratio','Smaller detour in Puxi_Jingan_West','Travelled / straight-line distance\\nHigher = more detour')]:\n                arrays=[legs.loc[legs.region_id.eq(name),column].to_numpy() for name in order]\n                boxes=axis.boxplot(arrays,tick_labels=labels,patch_artist=True,showfliers=True)\n                for box,color in zip(boxes['boxes'],['#3976a8','#c77c3d']):\n                    box.set_facecolor(color);box.set_alpha(.35)\n                for i,name in enumerate(order,1):\n                    mean=summary.loc[name,column]\n                    axis.scatter(i,mean,marker='D',s=48,color='black',zorder=4,label='Mean' if i==1 else None)\n                    axis.annotate(f'Mean {mean:.2f}',(i,mean),xytext=(14,5),textcoords='offset points',fontsize=9,\n                        bbox=dict(facecolor='white',alpha=.8,edgecolor='none',pad=1))\n                axis.set(title=title,ylabel=ylabel)\n                axis.legend(loc='upper left',frameon=False,fontsize=9)\n                axis.grid(axis='y',alpha=.15)\n                axis.margins(y=.18)\n            difference=summary.loc[order[1],'route_directness_ratio']/summary.loc[order[0],'route_directness_ratio']-1\n            axes[1].set_title(f'Puxi_Huangpu_East: {difference:.1%} higher mean ratio',fontsize=11)\n            axes[1].set_ylim(.9,3.25)\n            outside=int(legs.route_directness_ratio.gt(3.25).sum())\n            axes[1].text(.03,.88,f'Main view: 0.9–3.25\\n{outside} higher outlier shown in inset',\n                transform=axes[1].transAxes,fontsize=8,va='top')\n            inset=axes[1].inset_axes([.07,.43,.37,.34])\n            inset.boxplot([legs.loc[legs.region_id.eq(name),'route_directness_ratio'] for name in order],\n                tick_labels=['Jingan_West','Huangpu_East'],showfliers=True)\n            inset.set_title('Full range (all legs)',fontsize=8)\n            inset.tick_params(labelsize=6)\n            inset.grid(axis='y',alpha=.15)\n            value = {'region_means': summary.to_dict('index'), 'region_role_mapping': extra['regions'][['region_id', 'comparison_role']].to_dict('records')}\n        elif index == 2:\n            summary = legs.groupby(['region_id', 'period'])[['route_directness_ratio', 'primary_share', 'residential_share']].mean()\n            timed=legs.copy()\n            stamp=pd.to_datetime(timed.departure_time,format='mixed')\n            timed['departure_hour']=(stamp-stamp.dt.normalize()).dt.total_seconds()/3600\n            centers=np.arange(np.ceil(timed.departure_hour.min()*4)/4,np.floor(timed.departure_hour.max()*4)/4+.01,.25)\n            curves={}\n            plt.close(fig)\n            fig,ax=plt.subplots(figsize=(10,6),constrained_layout=True)\n            inset=ax.inset_axes([.08,.48,.34,.26])\n            for name,color in [('Puxi_Jingan_West','#3976a8'),('Puxi_Huangpu_East','#c77c3d')]:\n                group=timed[timed.region_id.eq(name)]\n                ax.scatter(group.departure_hour,group.route_directness_ratio,s=15,color=color,alpha=.25)\n                inset.scatter(group.departure_hour,group.route_directness_ratio,s=5,color=color,alpha=.4)\n                y=[];records=[]\n                for center in centers:\n                    window=group[group.departure_hour.between(center-.5,center+.5)]\n                    valid=len(window)>=8 and window.service_date.nunique()>=3\n                    y.append(float(window.route_directness_ratio.median()) if valid else np.nan)\n                    if valid:records.append(dict(center_hour=float(center),median_route_ratio=y[-1],legs=len(window),days=int(window.service_date.nunique())))\n                curves[name]=records\n                ax.plot(centers,y,'o-',color=color,linewidth=2.3,markersize=4,label=name)\n                inset.plot(centers,y,color=color,linewidth=1)\n            ticks=np.arange(8,np.ceil(timed.departure_hour.max())+.1,.5)\n            labels=[f'{int(h):02}:{int(round((h%1)*60)):02}' for h in ticks]\n            ax.set(xticks=ticks,xticklabels=labels,\n                ylabel='Travelled / straight-line distance (higher = more detour)',\n                xlabel='Actual departure time (Shanghai local time; 14 days overlaid)',\n                title='Regional detour trends over actual departure time\\nDots: individual legs; lines: 60-min rolling medians, every 15 min',\n                ylim=(.9,3.25),xlim=(7.9,float(timed.departure_hour.max())+.1))\n            inset.set(title='Full range (all legs)',xlim=ax.get_xlim(),ylim=(.9,float(timed.route_directness_ratio.max())+.5))\n            inset.title.set_fontsize(8);inset.tick_params(labelsize=7)\n            inset.set_xticks([8,10,12],['08:00','10:00','12:00'])\n            ax.text(.03,.85,f'Main view: ratio 0.9–3.25; {int(timed.route_directness_ratio.gt(3.25).sum())} higher outlier in inset\\nTrend requires ≥8 legs from ≥3 days per window; no interpolation',\n                transform=ax.transAxes,fontsize=8,va='top')\n            ax.legend(loc='upper left',frameon=False)\n            value = {'region_period_means': summary.reset_index().to_dict('records'),\n                'period_means_note':'Legacy phase summary; plot uses actual departure times, not phase labels.',\n                'rolling_window_minutes':60,'evaluation_interval_minutes':15,\n                'minimum_legs':8,'minimum_days':3,'regional_clock_time_trends':curves}\n        else:\n            values = {}\n            plt.close(fig)\n            fig,axes=plt.subplots(1,2,figsize=(12,5.8),sharex=True,sharey=True,constrained_layout=True);ax=axes[0]\n            for axis,name,color in zip(axes,['Puxi_Jingan_West','Puxi_Huangpu_East'],['#3976a8','#c77c3d']):\n                q=legs[legs.region_id.eq(name)]\n                axis.scatter(q.oneway_share*100,q.route_directness_ratio,color=color,s=24,alpha=.5)\n                values[name] = {'pearson': q.oneway_share.corr(q.route_directness_ratio),\n                    'spearman': q.oneway_share.corr(q.route_directness_ratio, method='spearman'),\n                    'legs': len(q), 'oneway_share_unique_values': q.oneway_share.nunique(),\n                    'route_ratio_unique_values': q.route_directness_ratio.nunique()}\n                axis.set(title=f'{name}\\nPearson r = {values[name][\"pearson\"]:.2f}; Spearman ρ = {values[name][\"spearman\"]:.2f}',\n                    xlabel='One-way-road share of travelled distance (%)',xlim=(-3,103),ylim=(.9,3.25))\n                axis.text(.03,.96,f'n = {len(q)}; main view: ratio 0.9–3.25\\n{int(q.route_directness_ratio.gt(3.25).sum())} higher outlier(s); full range below',\n                    transform=axis.transAxes,va='top',fontsize=8)\n                inset=axis.inset_axes([.08,.47,.38,.34])\n                inset.scatter(q.oneway_share*100,q.route_directness_ratio,color=color,s=5,alpha=.5)\n                inset.set(xlim=(-3,103),ylim=(.9,float(legs.route_directness_ratio.max())+.5),title='Full range (all legs)')\n                inset.title.set_fontsize(8);inset.tick_params(labelsize=6)\n                axis.grid(alpha=.15)\n            axes[0].set_ylabel('Travelled / straight-line distance\\nHigher = more detour')\n            axes[1].tick_params(labelleft=True)\n            value = {'within_region_associations': values}\n    ax.grid(alpha=.15)\n    return json.loads(json.dumps(value, default=lambda x: x.item()))\n\nevidence_value = evidence(23, 3, df, user_df, legs, visits, extra)\nprint(evidence_value)\nplt.show()",
  "difficulty_level": 5,
  "difficulty_label": "high",
  "difficulty_version": "3.5",
  "difficulty_is_provisional": false,
  "difficulty_field_dependencies": [
    "fclass",
    "completion_time",
    "gps_time",
    "gps_position",
    "courier_id"
  ]
}
```

### **Question 4**: Does the roughly 1.97 exception retain normal speed and little one-way exposure relative to pure one-way routes averaging about 2.01 in route directness?

#### Analysis action

Interpret the constructed descriptive association, not an estimated causal effect.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

legs = pd.read_csv("movement_legs.csv")
visits = pd.read_csv("aoi_visits.csv")
extra = {}
extra["aois"] = pd.read_csv("aois.csv")


def evidence(case, index, df, user_df, legs, visits, extra):
    import json
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(8.4, 5.0), constrained_layout=True)
    moving = user_df[user_df.activity_state.eq("travelling")].copy()
    done = df[df.status.eq("completed")].copy()
    value = {}
    if case == 21:
        order = ["primary", "secondary", "tertiary", "residential"]
        moving = moving.merge(
            legs[["leg_id", "period"]], on="leg_id", validate="many_to_one"
        )
        if index == 1:
            summary = (
                moving.groupby("fclass")[["step_distance_km", "step_duration_s"]]
                .sum()
                .reindex(order)
            )
            summary["movement_speed_kmh"] = (
                summary.step_distance_km / summary.step_duration_s * 3600
            )
            summary["completions"] = done.delivery_fclass.value_counts().reindex(
                order, fill_value=0
            )
            summary["completions_per_km"] = (
                summary.completions / summary.step_distance_km
            )
            ax.bar(
                order,
                summary.movement_speed_kmh,
                color="#5c8ea7",
                alpha=0.8,
                label="Movement speed",
            )
            twin = ax.twinx()
            twin.plot(order, summary.completions_per_km, "o-", color="#bd7843")
            ax.set(ylabel="Distance / moving time (km/h)", xlabel="Native OSM fclass")
            twin.set_ylabel("Delivery completions per travelled km", color="#bd7843")
            value = {"road_class_summary": summary.to_dict("index")}
        elif index == 2:
            summary = (
                moving.groupby(["period", "fclass"])
                .step_distance_km.sum()
                .unstack(fill_value=0)
            )
            summary = summary.div(summary.sum(axis=1), axis=0).reindex(
                ["early", "late"]
            )
            summary[order].plot.bar(stacked=True, ax=ax, rot=0)
            ax.set(
                ylabel="Share of all moving distance",
                xlabel="First / second half of scheduled delivery stops",
            )
            tasks = done.merge(
                legs[["leg_id", "period"]], left_on="arrival_leg_id", right_on="leg_id"
            )
            value = {
                "distance_shares": summary.to_dict("index"),
                "completion_counts": tasks.period.value_counts().to_dict(),
            }
        else:
            # Smooth spatial AND temporal event kernel, independently measured
            # from task-completion positions/times rather than a generator label.
            frames = []
            for tid, q in moving.groupby("trajectory_id"):
                t = done[done.trajectory_id.eq(tid)]
                x = q[["lng", "lat"]].to_numpy() * [95.2, 111]
                y = t[["delivery_lng", "delivery_lat"]].to_numpy() * [95.2, 111]
                ds = np.linalg.norm(x[:, None] - y[None, :], axis=2)
                a = (
                    pd.to_datetime(q.gps_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                b = (
                    pd.to_datetime(t.completion_time, format="mixed")
                    .astype("int64")
                    .to_numpy()
                )
                dt = abs(a[:, None] - b[None, :]) / 6e10
                q = q.copy()
                q["nearby_activity"] = (np.exp(-ds / 0.30) * np.exp(-dt / 30)).sum(
                    axis=1
                )
                frames.append(q)
            points = pd.concat(frames)
            values = {}
            for cls in order:
                q = points[points.fclass.eq(cls)]
                bins = pd.qcut(q.nearby_activity, 5, duplicates="drop")
                med = q.groupby(bins, observed=True)[
                    ["nearby_activity", "speed_kmh"]
                ].median()
                line = ax.plot(med.nearby_activity, med.speed_kmh, "o-", label=cls)[0]
                sample = q.sample(min(350, len(q)), random_state=17)
                ax.scatter(
                    sample.nearby_activity,
                    sample.speed_kmh,
                    s=6,
                    alpha=0.10,
                    color=line.get_color(),
                )
                values[cls] = {
                    "points": len(q),
                    "spearman": q.nearby_activity.corr(q.speed_kmh, method="spearman"),
                    "low_to_high_activity_speed_drop_kmh": (
                        med.speed_kmh.iloc[0] - med.speed_kmh.iloc[-1]
                    ),
                    "quintile_medians": med.to_dict("list"),
                }
            ax.set(
                xlabel=(
                    "Nearby completion intensity (300 m / 30 min exponential kernel)"
                ),
                ylabel="Median moving speed (km/h)",
            )
            ax.legend()
            value = {"within_class_evidence": values}
    elif case == 22:
        if index == 1:
            counts = (
                done.groupby(["service_date", "aoi_id"]).size().unstack(fill_value=0)
            )
            span = user_df.groupby("trajectory_id").gps_time.agg(["min", "max"])
            duration = (
                pd.to_datetime(span["max"], format="mixed")
                - pd.to_datetime(span["min"], format="mixed")
            ).dt.total_seconds()
            shares = (
                visits.groupby(["trajectory_id", "aoi_id"])
                .duration_s.sum()
                .unstack(fill_value=0)
                .div(duration, axis=0)
            )
            summary = pd.DataFrame(
                {
                    "mean_deliveries_per_day": counts.mean(),
                    "mean_work_time_share": shares.mean(),
                }
            )
            summary.plot.bar(
                y="mean_deliveries_per_day", ax=ax, rot=0, legend=False, color="#608ea4"
            )
            ax.set(
                ylabel="Completed deliveries per observed day", xlabel="Constructed AOI"
            )
            twin = ax.twinx()
            twin.plot(
                np.arange(len(summary)),
                summary.mean_work_time_share * 100,
                "o-",
                color="#bd7843",
            )
            twin.set_ylabel("Mean observed workday time (%)")
            value = {
                "aoi_summary": summary.to_dict("index"),
                "observed_days": counts.shape[0],
                "regular_destination_fclasses": sorted(
                    done[done.regular_aoi.eq("yes")].delivery_fclass.unique().tolist()
                ),
                "fixed_receipt_sites": (
                    extra["receipt_events"]
                    .groupby("aoi_id")[["lng", "lat"]]
                    .nunique()
                    .to_dict("index")
                ),
            }
        elif index == 2:
            q = legs[legs.purpose.eq("delivery")]
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.4), constrained_layout=True)
            ax = axes[0]
            ax.boxplot(
                [q[q.period.eq(p)].route_directness_ratio for p in ["early", "late"]],
                tick_labels=["Early", "Late"],
            )
            ax.set(
                xlabel="Delivery route phase",
                ylabel="Travelled / straight-line distance",
                title=(
                    "Later delivery routes are more direct\nLower ratio = less detour"
                ),
            )
            dates = sorted(q.service_date.unique())
            totals = moving.groupby("service_date")[
                ["step_distance_km", "step_duration_s"]
            ].sum()
            baseline = totals.step_distance_km / totals.step_duration_s * 3600
            late = q[q.period.eq("late")].copy()
            late["day_mean_movement_speed_kmh"] = late.service_date.map(baseline)
            late["difference_from_day_mean_kmh"] = (
                late.mean_speed_kmh - late.day_mean_movement_speed_kmh
            )
            above = late.difference_from_day_mean_kmh.gt(0)
            x = late.service_date.map({day: i for i, day in enumerate(dates)}).to_numpy(
                dtype=float
            )
            x += np.where(late.groupby("service_date").cumcount().eq(0), -0.12, 0.12)
            axes[1].axhline(
                0,
                color="#555555",
                linewidth=1,
                linestyle="--",
                label="Same as day mean",
            )
            for mask, color, label in [
                (above, "#3976a8", "Above day mean"),
                (~above, "#c77c3d", "At or below day mean"),
            ]:
                axes[1].scatter(
                    x[mask],
                    late.loc[mask, "difference_from_day_mean_kmh"],
                    color=color,
                    s=48,
                    label=f"{label}: {int(mask.sum())}/{len(late)} ({mask.mean():.1%})",
                )
            axes[1].set(
                xticks=np.arange(len(dates)),
                xticklabels=[day[5:] for day in dates],
                xlabel="Workday — each point is a delivery leg as time drew to a close",
                ylabel="Leg speed − day mean movement speed (km/h)",
                title=(
                    "As time drew to a close:\nMore direct does not guarantee"
                    " above-average speed"
                ),
            )
            axes[1].margins(y=0.30)
            axes[1].tick_params(axis="x", rotation=45, labelsize=8)
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "period_means": (
                    q.groupby("period")[["route_directness_ratio", "mean_speed_kmh"]]
                    .mean()
                    .to_dict("index")
                ),
                "day_mean_definition": (
                    "All moving GPS distance / all moving GPS time, including return"
                    " legs and excluding stops"
                ),
                "daily_mean_movement_speed_kmh": baseline.to_dict(),
                "late_delivery_legs": len(late),
                "late_legs_above_day_mean": int(above.sum()),
                "late_share_above_day_mean": float(above.mean()),
                "late_leg_comparison": (
                    late[
                        [
                            "leg_id",
                            "service_date",
                            "mean_speed_kmh",
                            "day_mean_movement_speed_kmh",
                            "difference_from_day_mean_kmh",
                        ]
                    ].to_dict("records")
                ),
            }
        elif index == 3:
            q = done.copy()
            stamp = pd.to_datetime(q.completion_time, format="mixed")
            q["completion_hour"] = (
                stamp.dt.hour + stamp.dt.minute / 60 + stamp.dt.second / 3600
            )
            q["window_hours"] = (
                pd.to_datetime(q.time_window_end) - pd.to_datetime(q.time_window_start)
            ).dt.total_seconds() / 3600
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13.5, 5.5), constrained_layout=True)
            ax = axes[0]
            groups = [
                q[q.regular_aoi.eq("no") & q.window_hours.lt(9)],
                q[q.regular_aoi.eq("no") & q.window_hours.ge(9)],
                q[q.regular_aoi.eq("yes") & q.window_hours.ge(9)],
            ]
            labels = [
                "Non-regular AOIs\nWindow <9 h",
                "Non-regular AOIs\nWindow ≥9 h",
                "Regular AOIs (A/B)\nWindow ≥9 h",
            ]
            colors = ["#4f846c", "#3976a8", "#c77c3d"]
            boxes = ax.boxplot(
                [g.completion_hour for g in groups],
                tick_labels=labels,
                patch_artist=True,
            )
            for box, color in zip(boxes["boxes"], colors):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, g in enumerate(groups, 1):
                median = g.completion_hour.median()
                h = int(median)
                minute = int(round((median - h) * 60))
                ax.text(
                    i,
                    17.8,
                    f"n={len(g)}\nMedian {h:02}:{minute:02}",
                    ha="center",
                    fontsize=9,
                )
            ax.set(
                ylabel="Task completion time (Shanghai)",
                title="Both non-regular window groups finish earlier",
                ylim=(7.5, 18.5),
            )
            ax.yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            ax.tick_params(axis="x", labelsize=8)
            for g, label, color, marker in zip(groups, labels, colors, ["o", "s", "^"]):
                axes[1].scatter(
                    g.window_hours,
                    g.completion_hour,
                    s=16,
                    alpha=0.45,
                    color=color,
                    marker=marker,
                    label=label.replace("\n", "; "),
                )
            axes[1].set(
                xlabel="Service-window duration (hours)",
                ylabel="Task completion time (Shanghai)",
                title="Wide windows alone do not explain later completion",
                ylim=(7.5, 19.5),
            )
            axes[1].yaxis.set_major_formatter(lambda v, _: f"{int(v):02}:00")
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(alpha=0.15)
            nvisits = visits.groupby(["service_date", "aoi_id"]).size()
            value = {
                "group_medians": (
                    q.groupby("regular_aoi")[["completion_hour", "window_hours"]]
                    .median()
                    .to_dict("index")
                ),
                "regular_aoi_visits_per_day": (
                    nvisits.unstack(fill_value=0)[["A", "B"]].to_dict("list")
                ),
                "irregular_by_window_median_hour": (
                    q[q.regular_aoi.eq("no")]
                    .assign(
                        window_group=lambda d: np.where(
                            d.window_hours < 9, "Narrower (<9h)", "Wide (9h+)"
                        )
                    )
                    .groupby("window_group")
                    .completion_hour.median()
                    .to_dict()
                ),
                "plotted_groups": {
                    label.replace("\n", "; "): {
                        "tasks": len(g),
                        "median_completion_hour": float(g.completion_hour.median()),
                        "window_hours_min": float(g.window_hours.min()),
                        "window_hours_max": float(g.window_hours.max()),
                    }
                    for label, g in zip(labels, groups)
                },
                "uncompleted_tasks_excluded": int(df.status.ne("completed").sum()),
            }
        else:
            irregular = df[df.regular_aoi.eq("no")]
            summary = (
                irregular.groupby(["service_date", "status"])
                .size()
                .unstack(fill_value=0)
            )
            summary.plot.bar(stacked=True, ax=ax, rot=50)
            ax.set(ylabel="Available irregular-AOI tasks", xlabel="Workday")
            pending = irregular[irregular.status.eq("pending")]
            dates = pending.service_date.unique().tolist()
            min_dist = {}
            for day in dates:
                gps = user_df[user_df.service_date.eq(day)]
                destinations = pending[pending.service_date.eq(day)][
                    ["aoi_id", "delivery_lng", "delivery_lat"]
                ].drop_duplicates()
                for r in destinations.itertuples():
                    ds = np.linalg.norm(
                        (
                            gps[["lng", "lat"]].to_numpy()
                            - [r.delivery_lng, r.delivery_lat]
                        )
                        * [95200, 111000],
                        axis=1,
                    )
                    min_dist[r.aoi_id] = float(ds.min())
            value = {
                "pending_dates": dates,
                "pending_tasks": len(pending),
                "minimum_GPS_distance_to_unvisited_AOI_m": min_dist,
                "anomaly_day_irregular_visit_count": int(
                    len(
                        visits[
                            visits.service_date.isin(dates)
                            & ~visits.aoi_id.isin(["A", "B"])
                        ]
                    )
                ),
            }
    elif case == 23:
        q = legs[legs.purpose.eq("delivery")].copy()
        # Initial departure is not between two completed task locations.
        q = q[q.groupby("trajectory_id").cumcount().gt(0)]
        if index == 1:
            q = q[q.period.ne("exception") & q.separation_km.between(0.8, 1.3)]
            q["oneway_group"] = np.where(
                q.oneway_share >= 0.55,
                "High (55%+)",
                np.where(q.oneway_share <= 0.25, "Low (0–25%)", "Middle"),
            )
            q = q[q.oneway_group.ne("Middle")]
            summary = q.groupby("oneway_group")[
                ["separation_km", "route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            high, low = summary.loc["High (55%+)"], summary.loc["Low (0–25%)"]
            relative = pd.Series(
                {
                    "Route directness": (
                        (high.route_directness_ratio / low.route_directness_ratio - 1)
                        * 100
                    ),
                    "Average speed": (
                        (high.mean_speed_kmh / low.mean_speed_kmh - 1) * 100
                    ),
                }
            )
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), constrained_layout=True)
            ax = axes[0]
            for name, s in q.groupby("oneway_group"):
                ax.scatter(
                    s.separation_km, s.route_directness_ratio, label=name, alpha=0.6
                )
            ax.set(
                xlabel="Task endpoint separation (km)",
                ylabel="Travelled / straight-line distance",
                title="Matched endpoint separations (0.8–1.3 km)",
            )
            ax.legend()
            bars = axes[1].bar(
                relative.index,
                relative.values,
                color=["#3976a8", "#c77c3d"],
                alpha=0.82,
            )
            axes[1].axhline(0, color="#666666", linewidth=1)
            axes[1].bar_label(
                bars,
                labels=[
                    (
                        f"+{relative.iloc[0]:.1f}%\n{low.route_directness_ratio:.2f} →"
                        f" {high.route_directness_ratio:.2f}"
                    ),
                    (
                        f"+{relative.iloc[1]:.1f}%\n{low.mean_speed_kmh:.1f} →"
                        f" {high.mean_speed_kmh:.1f} km/h"
                    ),
                ],
                padding=5,
                fontsize=10,
            )
            axes[1].set(
                ylabel="High one-way group relative to low group (%)",
                title="Average speed differs less than route directness",
                ylim=(0, max(relative) * 1.42),
            )
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "matched_separation_range_km": [0.8, 1.3],
                "group_means": summary.to_dict("index"),
                "high_vs_low_relative_difference_percent": relative.to_dict(),
                "within_period_group_means": (
                    q.groupby(["period", "oneway_group"])[
                        ["route_directness_ratio", "mean_speed_kmh"]
                    ]
                    .mean()
                    .reset_index()
                    .to_dict("records")
                ),
            }
        elif index == 2:
            means = (
                q[q.period.ne("exception")]
                .groupby("period")[["mean_speed_kmh", "route_directness_ratio"]]
                .mean()
                .reindex(["early", "midday", "late"])
            )
            stamp = pd.to_datetime(q.departure_time, format="mixed")
            q["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.floor(q.departure_hour.min() * 6) / 6,
                np.ceil(q.departure_hour.max() * 6) / 6 + 1 / 12,
                1 / 6,
            )
            windows = []
            for center in centers:
                subset = q[q.departure_hour.between(center - 0.5, center + 0.5)]
                windows.append(
                    {
                        "center_hour": float(center),
                        "legs": len(subset),
                        **{
                            column: (
                                float(subset[column].median())
                                if len(subset) >= 8
                                else None
                            )
                            for column in ["mean_speed_kmh", "route_directness_ratio"]
                        },
                    }
                )
            trend = pd.DataFrame(windows)
            plt.close(fig)
            fig, axes = plt.subplots(
                2, 1, figsize=(9.5, 7), sharex=True, constrained_layout=True
            )
            fig.suptitle("Speed and detour over actual departure time", fontsize=15)
            for axis, column, color, title in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "#3976a8",
                    "Movement speed of each task-to-task leg",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "#c77c3d",
                    "Route detour of the same legs at the same times",
                ),
            ]:
                normal = q[q.period.ne("exception")]
                exception = q[q.period.eq("exception")]
                axis.scatter(
                    normal.departure_hour,
                    normal[column],
                    s=22,
                    color=color,
                    alpha=0.45,
                    label=(
                        f"Task-to-task legs ({len(normal)};"
                        f" {normal.service_date.nunique()} days)"
                    ),
                )
                axis.scatter(
                    exception.departure_hour,
                    exception[column],
                    s=110,
                    color="#8c405f",
                    marker="*",
                    label="Exceptional leg (insight 4)",
                    zorder=4,
                )
                axis.plot(
                    trend.center_hour,
                    trend[column],
                    color=color,
                    linewidth=2.5,
                    label="60-min rolling median (≥8 legs)",
                )
                axis.set_title(title, fontsize=11)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(x=0.03, y=0.30)
                axis.legend(loc="upper left", fontsize=8, frameon=False)
            axes[0].set_ylabel("Mean movement speed (km/h)")
            axes[1].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            ticks = np.arange(
                np.floor(q.departure_hour.min()), np.ceil(q.departure_hour.max()) + 1
            )
            axes[1].set(
                xticks=ticks,
                xticklabels=[f"{int(h):02}:00" for h in ticks],
                xlabel="Actual leg departure time (Shanghai local time; days overlaid)",
            )
            value = {
                "period_means": means.to_dict("index"),
                "period_means_note": (
                    "Legacy phase aggregates retained for compatibility; not used to"
                    " draw this chart."
                ),
                "plotted_task_to_task_legs": len(q),
                "exception_legs_included": int(q.period.eq("exception").sum()),
                "rolling_window_minutes": 60,
                "minimum_legs_per_window": 8,
                "clock_time_rolling_medians": [
                    window for window in windows if window["legs"] >= 8
                ],
                "omitted_window_centers_hour": [
                    window["center_hour"] for window in windows if window["legs"] < 8
                ],
            }
        elif index == 3:
            p = (
                q[q.period.ne("exception")]
                .groupby(["service_date", "period"])
                .agg(
                    detour_sd=("route_directness_ratio", "std"),
                    speed_sd=("mean_speed_kmh", "std"),
                    transitions=("fclass_transitions", "sum"),
                    distance=("distance_km", "sum"),
                )
            )
            p["transitions_per_km"] = p.transitions / p.distance
            cut = p.transitions_per_km.median()
            p["transition_group"] = np.where(
                p.transitions_per_km > cut, "Frequent", "Less frequent"
            )
            pearson = p.transitions_per_km.corr(p.detour_sd)
            spearman = p.transitions_per_km.corr(p.detour_sd, method="spearman")
            group_means = p.groupby("transition_group")[
                ["detour_sd", "speed_sd", "transitions_per_km"]
            ].mean()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12.5, 5.5), constrained_layout=True)
            ax = axes[0]
            ax.scatter(
                p.transitions_per_km,
                p.detour_sd,
                color="#3976a8",
                s=43,
                alpha=0.72,
                label=f"Day × route phase (n={len(p)})",
            )
            slope, intercept = np.polyfit(p.transitions_per_km, p.detour_sd, 1)
            line = np.linspace(
                p.transitions_per_km.min(), p.transitions_per_km.max(), 100
            )
            ax.plot(
                line,
                slope * line + intercept,
                "--",
                color="#244d69",
                linewidth=2,
                label="Descriptive linear fit",
            )
            ax.axvline(
                cut,
                color="#777777",
                linestyle=":",
                linewidth=1.3,
                label=f"Median split: {cut:.2f}/km",
            )
            ax.set(
                xlabel="Native fclass transitions per travelled km",
                ylabel="Within-period route-ratio SD",
                title=(
                    "Detour variability increases with road-class switching\nPearson"
                    f" r={pearson:.2f}; Spearman ρ={spearman:.2f}"
                ),
            )
            ax.legend(loc="upper left", fontsize=8, frameon=False)
            ax.grid(alpha=0.15)
            order = ["Less frequent", "Frequent"]
            arrays = [p.loc[p.transition_group.eq(name), "speed_sd"] for name in order]
            boxes = axes[1].boxplot(
                arrays, tick_labels=order, patch_artist=True, showfliers=True
            )
            for box, color in zip(boxes["boxes"], ["#789bb0", "#d09a58"]):
                box.set_facecolor(color)
                box.set_alpha(0.45)
            for i, name in enumerate(order, 1):
                mean = group_means.loc[name, "speed_sd"]
                axes[1].scatter(
                    i,
                    mean,
                    marker="D",
                    s=45,
                    color="black",
                    zorder=4,
                    label="Mean" if i == 1 else None,
                )
                axes[1].annotate(
                    f"Mean {mean:.2f}",
                    (i, mean),
                    xytext=(8, 5),
                    textcoords="offset points",
                    fontsize=9,
                )
            axes[1].set(
                xlabel="Transition-frequency group (median split)",
                ylabel="Across-leg speed SD (km/h)",
                title=(
                    "Speed variability can remain similar\nIndividual periods still"
                    " vary"
                ),
            )
            axes[1].legend(loc="upper left", fontsize=8, frameon=False)
            axes[1].grid(axis="y", alpha=0.15)
            value = {
                "correlation_transition_frequency_detour_sd": pearson,
                "spearman_transition_frequency_detour_sd": spearman,
                "group_means": (
                    p.groupby("transition_group")[
                        ["detour_sd", "speed_sd", "transitions_per_km"]
                    ]
                    .mean()
                    .to_dict("index")
                ),
                "period_count": len(p),
            }
        else:
            normal = q[q.period.ne("exception")]
            exception = q[q.period.eq("exception")]
            pure = normal[np.isclose(normal.oneway_share, 1, atol=1e-10)]
            other = normal[~normal.index.isin(pure.index)]
            pure_mean = float(pure.route_directness_ratio.mean())
            ax.scatter(
                other.oneway_share * 100,
                other.route_directness_ratio,
                alpha=0.35,
                label="Other task legs",
            )
            ax.scatter(
                pure.oneway_share * 100,
                pure.route_directness_ratio,
                marker="D",
                s=55,
                color="#3976a8",
                label=f"Pure one-way routes (mean ratio {pure_mean:.2f})",
            )
            ax.scatter(
                exception.oneway_share * 100,
                exception.route_directness_ratio,
                marker="*",
                s=220,
                color="#c66643",
                label="Exceptional period",
            )
            ax.axhline(
                pure_mean, color="#3976a8", linestyle="--", linewidth=1.3, alpha=0.8
            )
            ax.set(
                xlabel="One-way-road distance (%)",
                ylabel="Travelled / straight-line distance",
            )
            ax.legend()
            value = {
                "exception": (
                    exception[
                        [
                            "leg_id",
                            "route_directness_ratio",
                            "oneway_share",
                            "mean_speed_kmh",
                            "distance_km",
                            "separation_km",
                        ]
                    ].to_dict("records")
                ),
                "normal_speed_10_90_percentiles": (
                    normal.mean_speed_kmh.quantile([0.1, 0.9]).tolist()
                ),
                "pure_oneway_reference": {
                    "definition": (
                        "delivery legs whose travelled distance is entirely on native"
                        " OSM one-way edges"
                    ),
                    "legs": int(len(pure)),
                    "mean_route_directness_ratio": pure_mean,
                    "route_directness_ratio_std": float(
                        pure.route_directness_ratio.std(ddof=0)
                    ),
                    "route_directness_ratios": pure.route_directness_ratio.tolist(),
                },
            }
    elif case == 24:
        if index == 1:
            q = legs[legs.purpose.str.startswith("road_condition")]
            q.boxplot(column="speed_std_kmh", by="purpose", ax=ax)
            fig.suptitle("")
            ax.set(
                xlabel="Road condition",
                ylabel="Short-term GPS speed SD (km/h)",
                title="Stable residential travel versus alternating road classes",
            )
            value = {
                "condition_means": (
                    q.groupby("purpose")[
                        [
                            "speed_std_kmh",
                            "residential_share",
                            "tertiary_share",
                            "fclass_transitions",
                        ]
                    ]
                    .mean()
                    .to_dict("index")
                )
            }
        else:
            q = extra["crossings"].merge(
                legs,
                left_on="crossing_leg_id",
                right_on="leg_id",
                validate="one_to_one",
            )
            low = moving[moving.speed_kmh < 12].groupby("leg_id").step_duration_s.sum()
            q["low_speed_seconds"] = q.leg_id.map(low).fillna(0)
            approach = legs.set_index("leg_id").mean_speed_kmh
            q["speed_drop_kmh"] = q.approach_leg_id.map(approach) - q.mean_speed_kmh
            if index == 2:
                # Fluctuation is within-crossing GPS speed SD, not variation
                # between the average speeds of different crossings.
                recomputed = moving.groupby("leg_id").speed_kmh.agg(
                    lambda s: float(np.std(s.to_numpy(), ddof=0))
                )
                np.testing.assert_allclose(
                    q.speed_std_kmh, q.leg_id.map(recomputed), atol=1e-8
                )
                plt.close(fig)
                fig, axes = plt.subplots(
                    1, 2, figsize=(12.5, 5.5), constrained_layout=True
                )
                ax = axes[0]
                structures = (
                    q[["structure_osm_way_id", "structure_type"]]
                    .drop_duplicates()
                    .sort_values("structure_type")
                )
                ticks = []
                labels = []
                for axis, column, title, ylabel in [
                    (
                        axes[0],
                        "low_speed_seconds",
                        "Longer low-speed time later in the workday",
                        "Time below 12 km/h per crossing (seconds)",
                    ),
                    (
                        axes[1],
                        "speed_std_kmh",
                        "Greater speed fluctuations later in the workday",
                        "Within-crossing GPS speed SD (km/h)",
                    ),
                ]:
                    for i, structure in enumerate(structures.itertuples()):
                        group = q[
                            q.structure_osm_way_id.eq(structure.structure_osm_way_id)
                        ]
                        paired = group.pivot(
                            index=["service_date", "approach_fclass"],
                            columns="period",
                            values=column,
                        )
                        assert paired[["early", "late"]].notna().all().all()
                        for row in paired.itertuples():
                            axis.plot(
                                [i * 3, i * 3 + 1],
                                [row.early, row.late],
                                color="#aab4bc",
                                alpha=0.5,
                                linewidth=0.8,
                            )
                        for j, (period, color) in enumerate(
                            [("early", "#3976a8"), ("late", "#c77c3d")]
                        ):
                            values = group.loc[group.period.eq(period), column]
                            axis.scatter(
                                np.full(len(values), i * 3 + j),
                                values,
                                s=24,
                                color=color,
                                alpha=0.7,
                                label=(
                                    period.capitalize() + " crossing"
                                    if i == 0
                                    else None
                                ),
                                zorder=3,
                            )
                            mean = values.mean()
                            axis.plot(
                                [i * 3 + j - 0.18, i * 3 + j + 0.18],
                                [mean, mean],
                                color="black",
                                linewidth=2,
                                label="Group mean" if i == 0 and j == 0 else None,
                                zorder=4,
                            )
                            axis.annotate(
                                f"{mean:.2f}",
                                (i * 3 + j, values.max()),
                                xytext=(0, 10),
                                textcoords="offset points",
                                ha="center",
                                fontweight="bold",
                                fontsize=10,
                            )
                        if axis is axes[0]:
                            ticks.extend([i * 3, i * 3 + 1])
                            labels.extend(
                                [
                                    f"Early\n{structure.structure_type.title()}",
                                    f"Late\n{structure.structure_type.title()}",
                                ]
                            )
                    axis.set(
                        xticks=ticks,
                        xticklabels=labels,
                        ylabel=ylabel,
                        title=title,
                        xlabel=(
                            "Paired within the same day, structure and approach class"
                        ),
                    )
                    axis.set_ylim(
                        0,
                        (
                            group[column].max() * 1.5
                            if column == "speed_std_kmh"
                            else q[column].max() * 1.5
                        ),
                    )
                    axis.legend(loc="upper left", fontsize=8, frameon=False)
                    axis.grid(axis="y", alpha=0.15)
                value = {
                    "structure_period_means": (
                        q.groupby(["structure_osm_way_id", "structure_type", "period"])[
                            ["low_speed_seconds", "speed_std_kmh", "mean_speed_kmh"]
                        ]
                        .mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                    "crossings": len(q),
                    "unique_structures": q.structure_osm_way_id.nunique(),
                }
            else:
                q.boxplot(column="speed_drop_kmh", by="approach_fclass", ax=ax)
                fig.suptitle("")
                ax.set(
                    xlabel="Dominant road class in last 800 m before structure",
                    ylabel="Approach minus crossing speed (km/h)",
                    title="Road approach conditions and crossing slowdowns",
                )
                value = {
                    "approach_means": (
                        q.groupby("approach_fclass")[
                            ["speed_drop_kmh", "approach_dominant_share"]
                        ]
                        .mean()
                        .to_dict("index")
                    ),
                    "within_structure_period_means": (
                        q.groupby(["structure_osm_way_id", "period", "approach_fclass"])
                        .speed_drop_kmh.mean()
                        .reset_index()
                        .to_dict("records")
                    ),
                }
    elif case == 25:
        if index == 1:
            summary = legs.groupby("region_id")[
                ["route_directness_ratio", "mean_speed_kmh"]
            ].mean()
            summary["delivery_volume"] = done.groupby("region_id").size()
            plt.close(fig)
            fig, axes = plt.subplots(1, 2, figsize=(12, 5.6), constrained_layout=True)
            ax = axes[0]
            order = ["Puxi_Jingan_West", "Puxi_Huangpu_East"]
            labels = [
                name + f'\n{int(summary.loc[name,"delivery_volume"])} deliveries'
                for name in order
            ]
            for axis, column, title, ylabel in [
                (
                    axes[0],
                    "mean_speed_kmh",
                    "Lower movement speed in Puxi_Jingan_West",
                    "Leg mean movement speed (km/h)",
                ),
                (
                    axes[1],
                    "route_directness_ratio",
                    "Smaller detour in Puxi_Jingan_West",
                    "Travelled / straight-line distance\nHigher = more detour",
                ),
            ]:
                arrays = [
                    legs.loc[legs.region_id.eq(name), column].to_numpy()
                    for name in order
                ]
                boxes = axis.boxplot(
                    arrays, tick_labels=labels, patch_artist=True, showfliers=True
                )
                for box, color in zip(boxes["boxes"], ["#3976a8", "#c77c3d"]):
                    box.set_facecolor(color)
                    box.set_alpha(0.35)
                for i, name in enumerate(order, 1):
                    mean = summary.loc[name, column]
                    axis.scatter(
                        i,
                        mean,
                        marker="D",
                        s=48,
                        color="black",
                        zorder=4,
                        label="Mean" if i == 1 else None,
                    )
                    axis.annotate(
                        f"Mean {mean:.2f}",
                        (i, mean),
                        xytext=(14, 5),
                        textcoords="offset points",
                        fontsize=9,
                        bbox=dict(
                            facecolor="white", alpha=0.8, edgecolor="none", pad=1
                        ),
                    )
                axis.set(title=title, ylabel=ylabel)
                axis.legend(loc="upper left", frameon=False, fontsize=9)
                axis.grid(axis="y", alpha=0.15)
                axis.margins(y=0.18)
            difference = (
                summary.loc[order[1], "route_directness_ratio"]
                / summary.loc[order[0], "route_directness_ratio"]
                - 1
            )
            axes[1].set_title(
                f"Puxi_Huangpu_East: {difference:.1%} higher mean ratio", fontsize=11
            )
            axes[1].set_ylim(0.9, 3.25)
            outside = int(legs.route_directness_ratio.gt(3.25).sum())
            axes[1].text(
                0.03,
                0.88,
                f"Main view: 0.9–3.25\n{outside} higher outlier shown in inset",
                transform=axes[1].transAxes,
                fontsize=8,
                va="top",
            )
            inset = axes[1].inset_axes([0.07, 0.43, 0.37, 0.34])
            inset.boxplot(
                [
                    legs.loc[legs.region_id.eq(name), "route_directness_ratio"]
                    for name in order
                ],
                tick_labels=["Jingan_West", "Huangpu_East"],
                showfliers=True,
            )
            inset.set_title("Full range (all legs)", fontsize=8)
            inset.tick_params(labelsize=6)
            inset.grid(axis="y", alpha=0.15)
            value = {
                "region_means": summary.to_dict("index"),
                "region_role_mapping": (
                    extra["regions"][["region_id", "comparison_role"]].to_dict(
                        "records"
                    )
                ),
            }
        elif index == 2:
            summary = legs.groupby(["region_id", "period"])[
                ["route_directness_ratio", "primary_share", "residential_share"]
            ].mean()
            timed = legs.copy()
            stamp = pd.to_datetime(timed.departure_time, format="mixed")
            timed["departure_hour"] = (
                stamp - stamp.dt.normalize()
            ).dt.total_seconds() / 3600
            centers = np.arange(
                np.ceil(timed.departure_hour.min() * 4) / 4,
                np.floor(timed.departure_hour.max() * 4) / 4 + 0.01,
                0.25,
            )
            curves = {}
            plt.close(fig)
            fig, ax = plt.subplots(figsize=(10, 6), constrained_layout=True)
            inset = ax.inset_axes([0.08, 0.48, 0.34, 0.26])
            for name, color in [
                ("Puxi_Jingan_West", "#3976a8"),
                ("Puxi_Huangpu_East", "#c77c3d"),
            ]:
                group = timed[timed.region_id.eq(name)]
                ax.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=15,
                    color=color,
                    alpha=0.25,
                )
                inset.scatter(
                    group.departure_hour,
                    group.route_directness_ratio,
                    s=5,
                    color=color,
                    alpha=0.4,
                )
                y = []
                records = []
                for center in centers:
                    window = group[
                        group.departure_hour.between(center - 0.5, center + 0.5)
                    ]
                    valid = len(window) >= 8 and window.service_date.nunique() >= 3
                    y.append(
                        float(window.route_directness_ratio.median())
                        if valid
                        else np.nan
                    )
                    if valid:
                        records.append(
                            dict(
                                center_hour=float(center),
                                median_route_ratio=y[-1],
                                legs=len(window),
                                days=int(window.service_date.nunique()),
                            )
                        )
                curves[name] = records
                ax.plot(
                    centers,
                    y,
                    "o-",
                    color=color,
                    linewidth=2.3,
                    markersize=4,
                    label=name,
                )
                inset.plot(centers, y, color=color, linewidth=1)
            ticks = np.arange(8, np.ceil(timed.departure_hour.max()) + 0.1, 0.5)
            labels = [f"{int(h):02}:{int(round((h%1)*60)):02}" for h in ticks]
            ax.set(
                xticks=ticks,
                xticklabels=labels,
                ylabel="Travelled / straight-line distance (higher = more detour)",
                xlabel="Actual departure time (Shanghai local time; 14 days overlaid)",
                title=(
                    "Regional detour trends over actual departure time\nDots:"
                    " individual legs; lines: 60-min rolling medians, every 15 min"
                ),
                ylim=(0.9, 3.25),
                xlim=(7.9, float(timed.departure_hour.max()) + 0.1),
            )
            inset.set(
                title="Full range (all legs)",
                xlim=ax.get_xlim(),
                ylim=(0.9, float(timed.route_directness_ratio.max()) + 0.5),
            )
            inset.title.set_fontsize(8)
            inset.tick_params(labelsize=7)
            inset.set_xticks([8, 10, 12], ["08:00", "10:00", "12:00"])
            ax.text(
                0.03,
                0.85,
                "Main view: ratio 0.9–3.25;"
                f" {int(timed.route_directness_ratio.gt(3.25).sum())} higher outlier in"
                " inset\nTrend requires ≥8 legs from ≥3 days per window; no"
                " interpolation",
                transform=ax.transAxes,
                fontsize=8,
                va="top",
            )
            ax.legend(loc="upper left", frameon=False)
            value = {
                "region_period_means": summary.reset_index().to_dict("records"),
                "period_means_note": (
                    "Legacy phase summary; plot uses actual departure times, not phase"
                    " labels."
                ),
                "rolling_window_minutes": 60,
                "evaluation_interval_minutes": 15,
                "minimum_legs": 8,
                "minimum_days": 3,
                "regional_clock_time_trends": curves,
            }
        else:
            values = {}
            plt.close(fig)
            fig, axes = plt.subplots(
                1,
                2,
                figsize=(12, 5.8),
                sharex=True,
                sharey=True,
                constrained_layout=True,
            )
            ax = axes[0]
            for axis, name, color in zip(
                axes, ["Puxi_Jingan_West", "Puxi_Huangpu_East"], ["#3976a8", "#c77c3d"]
            ):
                q = legs[legs.region_id.eq(name)]
                axis.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=24,
                    alpha=0.5,
                )
                values[name] = {
                    "pearson": q.oneway_share.corr(q.route_directness_ratio),
                    "spearman": q.oneway_share.corr(
                        q.route_directness_ratio, method="spearman"
                    ),
                    "legs": len(q),
                    "oneway_share_unique_values": q.oneway_share.nunique(),
                    "route_ratio_unique_values": q.route_directness_ratio.nunique(),
                }
                axis.set(
                    title=(
                        f'{name}\nPearson r = {values[name]["pearson"]:.2f}; Spearman ρ'
                        f' = {values[name]["spearman"]:.2f}'
                    ),
                    xlabel="One-way-road share of travelled distance (%)",
                    xlim=(-3, 103),
                    ylim=(0.9, 3.25),
                )
                axis.text(
                    0.03,
                    0.96,
                    f"n = {len(q)}; main view: ratio"
                    f" 0.9–3.25\n{int(q.route_directness_ratio.gt(3.25).sum())} higher"
                    " outlier(s); full range below",
                    transform=axis.transAxes,
                    va="top",
                    fontsize=8,
                )
                inset = axis.inset_axes([0.08, 0.47, 0.38, 0.34])
                inset.scatter(
                    q.oneway_share * 100,
                    q.route_directness_ratio,
                    color=color,
                    s=5,
                    alpha=0.5,
                )
                inset.set(
                    xlim=(-3, 103),
                    ylim=(0.9, float(legs.route_directness_ratio.max()) + 0.5),
                    title="Full range (all legs)",
                )
                inset.title.set_fontsize(8)
                inset.tick_params(labelsize=6)
                axis.grid(alpha=0.15)
            axes[0].set_ylabel(
                "Travelled / straight-line distance\nHigher = more detour"
            )
            axes[1].tick_params(labelleft=True)
            value = {"within_region_associations": values}
    ax.grid(alpha=0.15)
    return json.loads(json.dumps(value, default=lambda x: x.item()))


evidence_value = evidence(23, 4, df, user_df, legs, visits, extra)
print(evidence_value)
plt.close("all")
plt.close("all")
# Display the verified visualization once.
display(Image(filename="visualizations/insight_4.png"))
```

### Summary of Findings (Flag 23)

Among movements between similarly separated tasks, those using a higher share of one-way streets show larger detour ratios, while their average speeds differ less. Speed and route directness ratios evolve differently over the workday: the slowest periods are not always the most circuitous, and some high-detour periods retain relatively normal speeds. Periods with frequent transitions between different fclass categories show greater fluctuations in detour ratio, while movement speed can remain relatively stable. One route period shows a very high route directness ratio despite normal movement speed and limited one-way-road exposure, differing from the dominant road-related detour pattern.
